## 1. Fundamentos Teóricos

### 1.1 Oscilador de FitzHugh-Nagumo
O oscilador de FitzHugh-Nagumo é um modelo simplificado bidimensional da excitação neuronal, proposto por Richard FitzHugh (1961) e Nagumo et al. (1962). Ele reduz a complexidade do modelo de Hodgkin-Huxley a duas variáveis acopladas. A dinâmica é baseada nas seguintes premissas:

1. O sistema possui uma variável rápida (potencial de membrana) e uma variável lenta (recuperação).
2. A variável rápida segue uma dinâmica de excitação com não linearidade cúbica.
3. A variável lenta atua como um "freio", promovendo refratariedade e recuperação após o disparo.

### 1.2 Equações de FitzHugh-Nagumo
O sistema é descrito por um par de equações diferenciais de primeira ordem:

$$
\boxed{
\begin{cases}
\dfrac{dv}{dt} = v - \dfrac{v^3}{3} - w + R I_{ext} \\[10pt]
\dfrac{dw}{dt} = \epsilon (v + a - b w)
\end{cases}
}
$$

Onde:
- $v$ é o **potencial de membrana** (variável rápida, excitação)
- $w$ é a **variável de recuperação** (variável lenta, inativação)
- $R$ é a **resistência de acoplamento**
- $I_{ext}$ é a **corrente aplicada** (entrada externa)
- $\epsilon$ é o **parâmetro de separação de escalas** ($0 < \epsilon \ll 1$)
- $a$ e $b$ são **parâmetros positivos** que controlam a dinâmica da recuperação

O termo $v - v^3/3 - w$ representa a dinâmica rápida da excitação, com não linearidade cúbica que permite disparos do tipo "tudo-ou-nada". A variável $w$ atua como corrente de recuperação que inibe a excitação.

## 2. Pontos de Equilíbrio

### 2.1 Pontos Críticos
Os pontos de equilíbrio são obtidos fazendo $\dot{v} = 0$ e $\dot{w} = 0$:

$$
\begin{cases}
v - \dfrac{v^3}{3} - w + R I_{ext} = 0 \\[5pt]
\epsilon (v + a - b w) = 0
\end{cases}
$$

Da segunda equação: $w = \dfrac{v + a}{b}$

Substituindo na primeira:

$$
v - \dfrac{v^3}{3} - \dfrac{v + a}{b} + R I_{ext} = 0
$$

Isso resulta em uma **equação cúbica** em $v$, que pode ter **1 ou 3 pontos de equilíbrio** dependendo dos parâmetros:

$$
\boxed{\dfrac{v^3}{3} - \left(1 - \dfrac{1}{b}\right)v + \left(\dfrac{a}{b} - R I_{ext}\right) = 0}
$$

### 2.2 Estabilidade dos Pontos de Equilíbrio
A estabilidade é determinada pela matriz Jacobiana:

$$
J = \begin{bmatrix}
1 - v^2 & -1 \\
\epsilon & -\epsilon b
\end{bmatrix}
$$

O traço e o determinante são:

$$
\text{Tr}(J) = 1 - v^2 - \epsilon b
$$

$$
\det(J) = -\epsilon b (1 - v^2) + \epsilon = \epsilon [1 - b(1 - v^2)]
$$

- **Ponto de equilíbrio estável**: Se $\text{Tr}(J) < 0$ e $\det(J) > 0$
- **Ponto de equilíbrio instável**: Se $\text{Tr}(J) > 0$ ou $\det(J) < 0$
- **Ponto de sela**: Se $\det(J) < 0$
- **Bifurcação de Hopf**: Ocorre quando $\text{Tr}(J) = 0$ e $\det(J) > 0$

## 3. Solução Analítica

### 3.1 Comportamento Assintótico
O sistema de FitzHugh-Nagumo não possui solução analítica fechada. Para $\epsilon \ll 1$, o sistema apresenta **dinâmica de dois tempos**:

1. **Fase rápida** ($v$ varia rapidamente, $w$ quase constante)
2. **Fase lenta** ($w$ varia lentamente, $v$ segue a nuliclina $\dot{v} = 0$)

### 3.2 Ciclo Limite
Para certos valores de $I$, o sistema apresenta um **ciclo limite estável** (oscilações de relaxação), com dois regimes distintos:

1. **Regime lento**: A trajetória segue a nuliclina $\dot{v} = 0$ (ou seja, $w = v - v^3/3 + R I_{ext}$)
2. **Regime rápido**: Transições abruptas entre os ramos da nuliclina (disparos)

O ciclo limite é encontrado numericamente, pois não há solução analítica fechada.

## 4. Parâmetros do Sistema

### 4.1 Período de Oscilação
Para $\epsilon \ll 1$, o período é dominado pelo tempo de recuperação lenta e escala como:

$$
\boxed{T \propto \frac{1}{\epsilon}}
$$

Para valores típicos ($\epsilon = 0.08$, $a = 0.7$, $b = 0.8$), o período é da ordem de dezenas a centenas de unidades de tempo.

### 4.2 Frequência Angular
A frequência angular é aproximadamente:

$$
\boxed{\omega \propto \epsilon}
$$

### 4.3 Amplitude do Ciclo Limite
A amplitude do potencial de membrana é aproximadamente constante:

$$
\boxed{A_v \approx 2}
$$

A amplitude da variável de recuperação depende dos parâmetros:

$$
\boxed{A_w \approx \dfrac{2 + a}{b}}
$$

## 5. Constante de Movimento

### 5.1 Função de "Energia"
Podemos definir uma função de "energia" para o sistema (pseudo-energia):

$$
E(v,w) = \dfrac{v^2}{2} + \dfrac{w^2}{2}
$$

### 5.2 Variação da Energia
Ao longo de uma trajetória, a derivada total da energia é:

$$
\frac{dE}{dt} = v\dot{v} + w\dot{w} = v\left(v - \frac{v^3}{3} - w + R I_{ext}\right) + w\epsilon(v + a - b w)
$$

$$
\frac{dE}{dt} = v^2 - \frac{v^4}{3} - vw + RI_{ext}v + \epsilon w v + \epsilon a w - \epsilon b w^2
$$

A energia do sistema **não é conservada**. O sistema ganha e perde energia ao longo do ciclo, estabilizando-se em um valor médio correspondente ao ciclo limite.

### 5.3 Ciclo Limite
A energia se estabiliza em um valor médio constante correspondente ao ciclo limite:

$$
\boxed{E_{\text{limite}} \approx \dfrac{A_v^2}{2} + \dfrac{A_w^2}{2}}
$$

## 6. Sistema de Equações de Primeira Ordem

Para aplicação do **Runge-Kutta**, o sistema já está na forma de primeira ordem:

Definindo:
- $y_1 = v$ (potencial de membrana)
- $y_2 = w$ (variável de recuperação)

O sistema fica:

$$
\begin{cases}
\dot{y}_1 = y_1 - \dfrac{y_1^3}{3} - y_2 + R I_{ext} \\[10pt]
\dot{y}_2 = \epsilon (y_1 + a - b y_2)
\end{cases}
$$

## 7. Espaço de Fases

O **espaço de fases** é definido pelos eixos $(v, w)$, onde cada ponto representa o estado instantâneo do neurônio.

### 7.1 Trajetória no Espaço de Fases
Para o oscilador de FitzHugh-Nagumo com parâmetros que produzem oscilação, as trajetórias são **curvas fechadas** (ciclo limite) no espaço de fases:

- Para condições iniciais dentro do ciclo: trajetórias espiralam para fora
- Para condições iniciais fora do ciclo: trajetórias espiralam para dentro
- No ciclo limite: trajetórias periódicas (spikes neuronais)

### 7.2 Características das Órbitas
- **Forma aproximadamente oval**: Para $\epsilon$ moderado
- **Forma com transições abruptas (oscilação de relaxação)**: Para $\epsilon \ll 1$
- **Nuliclinas**:
  - Nuliclina de $v$: $w = v - v^3/3 + R I_{ext}$ (curva em forma de N)
  - Nuliclina de $w$: $w = (v + a)/b$ (reta inclinada)
- **Sentido**: Anti-horário (oscilações periódicas)
- **Amplitude**: $A_v \approx 2$ (ciclo limite)

### 7.3 Ciclo de Oscilação (para $\epsilon \ll 1$)
1. **Fase de disparo (subida rápida)**: $v$ sobe rapidamente de $-2$ para $2$ (ou de um valor negativo para positivo)
2. **Fase de refratariedade (descida lenta)**: $v$ cai lentamente enquanto $w$ aumenta
3. **Fase de recuperação (descida rápida)**: $v$ cai rapidamente de volta ao valor de repouso
4. **Fase de repouso (lenta)**: $v$ e $w$ retornam lentamente ao ponto de disparo

## 8. Método Runge-Kutta de 4ª Ordem (RK4)

Para resolver numericamente o sistema, utilizamos o método RK4:

Dado o sistema $\dot{\mathbf{y}} = \mathbf{f}(t, \mathbf{y})$, com passo $h = \Delta t$, temos como equações principais:

$$
\begin{aligned}
\mathbf{k}_1 &= \mathbf{f}(t_n, \mathbf{y}_n) \\
\mathbf{k}_2 &= \mathbf{f}\left(t_n + \frac{h}{2}, \mathbf{y}_n + \frac{h}{2}\mathbf{k}_1\right) \\
\mathbf{k}_3 &= \mathbf{f}\left(t_n + \frac{h}{2}, \mathbf{y}_n + \frac{h}{2}\mathbf{k}_2\right) \\
\mathbf{k}_4 &= \mathbf{f}\left(t_n + h, \mathbf{y}_n + h\mathbf{k}_3\right)
\end{aligned}
$$

$$
\boxed{\mathbf{y}_{n+1} = \mathbf{y}_n + \frac{h}{6}(\mathbf{k}_1 + 2\mathbf{k}_2 + 2\mathbf{k}_3 + \mathbf{k}_4)}
$$

## 8.1. O que é o RK4?

É uma fórmula que calcula a posição e velocidade no próximo instante de tempo utilizando 4 estimativas das derivadas dentro do intervalo.

## 8.2. O problema

O sistema de FitzHugh-Nagumo na forma de primeira ordem é:

$$
\begin{cases}
v' = v - \dfrac{v^3}{3} - w + R I_{ext} \\
w' = \epsilon (v + a - b w)
\end{cases}
$$

## 8.3. A fórmula

Para avançar de $(v_n, w_n)$ para $(v_{n+1}, w_{n+1})$ com passo $h$:

$$
\boxed{(v_{n+1}, w_{n+1}) = (v_n, w_n) + \frac{h}{6}(k_1 + 2k_2 + 2k_3 + k_4)}
$$

## 8.4. As variáveis

| k | Onde calcula | Como calcula |
|---|--------------|--------------|
| $k_1$ | no início | $\left(v_n - \dfrac{v_n^3}{3} - w_n + R I_{ext},\; \epsilon (v_n + a - b w_n)\right)$ |
| $k_2$ | no meio | $\left(\begin{array}{l} (v_n + \frac{h}{2}k_{1v}) - \frac{(v_n + \frac{h}{2}k_{1v})^3}{3} - (w_n + \frac{h}{2}k_{1w}) + R I_{ext},\; \epsilon ((v_n + \frac{h}{2}k_{1v}) + a - b(w_n + \frac{h}{2}k_{1w})) \end{array}\right)$ |
| $k_3$ | no meio | $\left(\begin{array}{l} (v_n + \frac{h}{2}k_{2v}) - \frac{(v_n + \frac{h}{2}k_{2v})^3}{3} - (w_n + \frac{h}{2}k_{2w}) + R I_{ext},\; \epsilon ((v_n + \frac{h}{2}k_{2v}) + a - b(w_n + \frac{h}{2}k_{2w})) \end{array}\right)$ |
| $k_4$ | no fim | $\left(\begin{array}{l} (v_n + h k_{3v}) - \frac{(v_n + h k_{3v})^3}{3} - (w_n + h k_{3w}) + R I_{ext},\; \epsilon ((v_n + h k_{3v}) + a - b(w_n + h k_{3w})) \end{array}\right)$ |

Média ponderada (1,2,2,1) dá mais peso ao meio do intervalo.

**Nota**: $k_{iv}$ e $k_{iw}$ representam as componentes de $k_i$ para potencial de membrana e variável de recuperação, respectivamente.

## 8.5. Exemplo numérico

Para $\epsilon = 0.08$, $a = 0.7$, $b = 0.8$, $I = 0.5$, $h = 0.1$, partindo de $v = 0.5$, $w = 0.0$:

**Cálculo dos k's:**

$$
\begin{aligned}
k_1 &= \left(0.5 - \frac{0.5^3}{3} - 0.0 + 0.5,\; 0.08(0.5 + 0.7 - 0.8 \cdot 0.0)\right) \\
&= (0.4583, 0.0960)
\end{aligned}
$$

$$
\begin{aligned}
k_2 &= \left(\begin{array}{l}
(0.5 + 0.05 \cdot 0.4583) - \frac{(0.5 + 0.05 \cdot 0.4583)^3}{3} - (0.0 + 0.05 \cdot 0.0960) + 0.5, \\
0.08((0.5 + 0.05 \cdot 0.4583) + 0.7 - 0.8(0.0 + 0.05 \cdot 0.0960))
\end{array}\right) \\
&= (0.4803, 0.0953)
\end{aligned}
$$

$$
\begin{aligned}
k_3 &= \left(\begin{array}{l}
(0.5 + 0.05 \cdot 0.4803) - \frac{(0.5 + 0.05 \cdot 0.4803)^3}{3} - (0.0 + 0.05 \cdot 0.0953) + 0.5, \\
0.08((0.5 + 0.05 \cdot 0.4803) + 0.7 - 0.8(0.0 + 0.05 \cdot 0.0953))
\end{array}\right) \\
&= (0.4804, 0.0953)
\end{aligned}
$$

$$
\begin{aligned}
k_4 &= \left(\begin{array}{l}
(0.5 + 0.1 \cdot 0.4804) - \frac{(0.5 + 0.1 \cdot 0.4804)^3}{3} - (0.0 + 0.1 \cdot 0.0953) + 0.5, \\
0.08((0.5 + 0.1 \cdot 0.4804) + 0.7 - 0.8(0.0 + 0.1 \cdot 0.0953))
\end{array}\right) \\
&= (0.5036, 0.0952)
\end{aligned}
$$

**Próximos estados:**

$$
\begin{aligned}
v_{n+1} &= 0.5 + \frac{0.1}{6} \cdot (0.4583 + 2 \cdot 0.4803 + 2 \cdot 0.4804 + 0.5036) = 0.5481 \\
w_{n+1} &= 0.0 + \frac{0.1}{6} \cdot (0.0960 + 2 \cdot 0.0953 + 2 \cdot 0.0953 + 0.0952) = 0.00954
\end{aligned}
$$

## 9. Ressalvas

### 9.1 Premissas do Modelo
1. **Duas variáveis apenas**: Simplificação do modelo de Hodgkin-Huxley (originalmente com 4 variáveis)
2. **Não linearidade cúbica**: Aproximação da soma das correntes iônicas rápidas
3. **Sistema autônomo**: O sistema não depende explicitamente do tempo (exceto se $I_{ext}$ variar)
4. **Parâmetros constantes**: $\epsilon$, $R$, $a$ e $b$ são fixos
5. **Corrente aplicada constante**: $I_{ext}$ é constante (ou variável no tempo, se forçado externamente)

### 9.2 Limitações do Modelo
- Em neurônios reais, a dinâmica iônica é mais complexa (canais de Na⁺, K⁺, Ca²⁺, etc.)
- O modelo não captura todos os tipos de disparos neuronais (ex.: bursting, adaptação)
- Parâmetros podem variar com a temperatura, concentração iônica, ou estado do neurônio
- Para $\epsilon$ muito pequeno, o sistema se torna stiff e requer métodos numéricos especiais
- O modelo não considera ruído sináptico ou efeitos estocásticos
- Não inclui a dinâmica dendrítica ou de axônios
- O ciclo limite é estável apenas para certas faixas de $I_{ext}$ e parâmetros

## Principais Equações

| Grandeza | Equação |
|----------|---------|
| Equações de FitzHugh-Nagumo | $\dot{v} = v - \dfrac{v^3}{3} - w + R I_{ext},\; \dot{w} = \epsilon (v + a - b w)$ |
| Pontos de equilíbrio | $v - \dfrac{v^3}{3} - \dfrac{v + a}{b} + R I_{ext} = 0,\; w = \dfrac{v + a}{b}$ |
| Período ($\epsilon \ll 1$) | $T \propto \dfrac{1}{\epsilon}$ |
| Amplitude do ciclo limite ($v$) | $A_v \approx 2$ |
| Amplitude do ciclo limite ($w$) | $A_w \approx \dfrac{2 + a}{b}$ |

## Equação a ser Resolvida com RK4

$$
\boxed{
\begin{cases}
\dfrac{dv}{dt} = v - \dfrac{v^3}{3} - w + R I_{ext} \\[10pt]
\dfrac{dw}{dt} = \epsilon (v + a - b w)
\end{cases}
}
$$
