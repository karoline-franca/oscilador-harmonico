## 1. Fundamentos Teóricos

### 1.1 Dinâmica de Populações
O modelo de Lotka-Volterra descreve a interação entre duas populações: **presas** e **predadores**. A dinâmica é baseada nas seguintes premissas:

1. Na ausência de predadores, as presas crescem exponencialmente
2. Na ausência de presas, os predadores morrem exponencialmente
3. A taxa de predação é proporcional ao encontro entre presas e predadores
4. A taxa de crescimento dos predadores é proporcional à quantidade de presas consumidas

### 1.2 Equações de Lotka-Volterra
O sistema é descrito por duas equações diferenciais acopladas:

$$
\boxed{\frac{dx}{dt} = a x - b x y}
$$

$$
\boxed{\frac{dy}{dt} = -c y + d x y}
$$

Onde:
- $x$ é a **população de presas**
- $y$ é a **população de predadores**
- $a$ é a **taxa de crescimento das presas** (1/t)
- $b$ é a **taxa de predação** (1/ind·t)
- $c$ é a **taxa de mortalidade dos predadores** (1/t)
- $d$ é a **eficiência de conversão** de presas novos predadores (1/ind·t)

O termo $bxy$ representa a taxa de encontro entre presas e predadores, enquanto $dxy$ representa a taxa de conversão de presas em predadores.

## 2. Pontos de Equilíbrio

### 2.1 Pontos Críticos
Os pontos de equilíbrio são obtidos fazendo $\frac{dx}{dt} = 0$ e $\frac{dy}{dt} = 0$:

**Ponto trivial:**
$$
\boxed{(x^*, y^*) = (0, 0)}
$$

**Ponto não-trivial (coexistência):**
$$
\boxed{(x^*, y^*) = \left(\frac{c}{d}, \frac{a}{b}\right)}
$$

### 2.2 Estabilidade dos Pontos de Equilíbrio
- **Ponto (0,0)**: Ponto de sela (instável)
- **Ponto (c/d, a/b)**: Centro (neutramente estável) - as trajetórias são órbitas fechadas

## 3. Solução Analítica

### 3.1 Constante de Movimento
O sistema de Lotka-Volterra possui uma **constante de movimento**:

$$
\boxed{H(x,y) = d x + b y - c \ln(x) - a \ln(y) = \text{constante}}
$$

Esta constante é invariante ao longo das trajetórias, caracterizando um sistema conservativo.

### 3.2 Solução Implícita
Embora não exista solução analítica explícita em termos de funções elementares, a solução é dada implicitamente por:

$$
\boxed{d x + b y - c \ln(x) - a \ln(y) = H_0}
$$

Onde $H_0$ é determinado pelas condições iniciais $(x_0, y_0)$.

## 4. Parâmetros do Sistema

### 4.1 Período de Oscilação
O período das oscilações (aproximação linear em torno do equilíbrio) é:

$$
\boxed{T \approx \frac{2\pi}{\sqrt{ac}}}
$$

### 4.2 Frequência Angular
A frequência angular das oscilações (aproximação linear) é:

$$
\boxed{\omega_0 \approx \sqrt{ac}}
$$

### 4.3 Valores Médios
As médias temporais das populações são exatamente os pontos de equilíbrio:

$$
\boxed{\langle x \rangle = \frac{c}{d}}
$$

$$
\boxed{\langle y \rangle = \frac{a}{b}}
$$

## 5. Constante de Movimento

### 5.1 Função de Lyapunov
A função $H(x,y)$ pode ser interpretada como uma "energia" do sistema:

$$
H(x,y) = d x + b y - c \ln(x) - a \ln(y)
$$

### 5.2 Variação da Constante de Movimento
Ao longo de uma trajetória, a derivada total de $H$ é zero:

$$
\frac{dH}{dt} = \frac{\partial H}{\partial x}\frac{dx}{dt} + \frac{\partial H}{\partial y}\frac{dy}{dt} = 0
$$

Isso confirma que $H$ é uma constante de movimento.

### 5.3 Constantes de Movimento

$$
\boxed{H(x(t), y(t)) = H(x_0, y_0) = \text{constante}}
$$

## 6. Sistema de Equações de Primeira Ordem

Para aplicação do **Runge-Kutta**, o sistema já está na forma de primeira ordem.

Definindo:
- $y_1 = x$ (presas)
- $y_2 = y$ (predadores)

O sistema fica:

$$
\begin{cases}
\dot{y}_1 = a y_1 - b y_1 y_2 \\[10pt]
\dot{y}_2 = -c y_2 + d y_1 y_2
\end{cases}
$$

## 7. Espaço de Fases

O **espaço de fases** é definido pelos eixos $(x, y)$, onde cada ponto representa o estado instantâneo do sistema.

### 7.1 Trajetória no Espaço de Fases
Para o oscilador de Lotka-Volterra, as trajetórias são **curvas fechadas** (órbitas periódicas) ao redor do ponto de equilíbrio não-trivial:

$$
\boxed{d x + b y - c \ln(x) - a \ln(y) = H_0}
$$

### 7.2 Características das Órbitas
- **Forma**: Curvas ovais deformadas (não são elipses exatas)
- **Centro**: Ponto de equilíbrio $(c/d, a/b)$
- **Sentido**: Anti-horário (as populações oscilam em ciclo)
- A amplitude das oscilações depende do valor de $H_0$ (condições iniciais)

### 7.3 Ciclo de Oscilação
1. **Presas aumentam** → Predadores comem mais
2. **Predadores aumentam** → Presas diminuem
3. **Presas diminuem** → Predadores morrem de fome
4. **Predadores diminuem** → Presas se recuperam
5. O ciclo recomeça

## 8. Método Runge-Kutta de 4ª Ordem (RK4)

Para resolver numericamente o sistema, utilizamos o método RK4:

Dado o sistema $\dot{\mathbf{y}} = \mathbf{f}(t, \mathbf{y})$, com passo $h = \Delta t$, temos como equações principais:

$$
\begin{aligned}
\mathbf{k}_1 &= \mathbf{f}(t_n, \mathbf{y}_n) \\
\mathbf{k}_2 &= \mathbf{f}\left(t_n + \frac{h}{2}, \mathbf{y}_n + \frac{h}{2}\mathbf{k}_1\right) \\
\mathbf{k}_3 &= \mathbf{f}\left(t_n + \frac{h}{2}, \mathbf{y}_n + \frac{h}{2}\mathbf{k}_2\right) \\
\mathbf{k}_4 &= \mathbf{f}\left(t_n + h, \mathbf{y}_n + h\mathbf{k}_3\right)
\end{aligned}
$$

$$
\boxed{\mathbf{y}_{n+1} = \mathbf{y}_n + \frac{h}{6}(\mathbf{k}_1 + 2\mathbf{k}_2 + 2\mathbf{k}_3 + \mathbf{k}_4)}
$$

## 8.1. O que é o RK4?

É uma fórmula que calcula as populações de presas e predadores no próximo instante de tempo utilizando 4 estimativas das derivadas dentro do intervalo.

## 8.2. O problema

O sistema de Lotka-Volterra já está na forma de primeira ordem:

$$
\begin{cases}
x' = a x - b x y \\
y' = -c y + d x y
\end{cases}
$$

## 8.3. A fórmula

Para avançar de $(x_n, y_n)$ para $(x_{n+1}, y_{n+1})$ com passo $h$:

$$
\boxed{(x_{n+1}, y_{n+1}) = (x_n, y_n) + \frac{h}{6}(k_1 + 2k_2 + 2k_3 + k_4)}
$$

## 8.4. As variáveis

| k | Onde calcula | Como calcula |
|---|--------------|--------------|
| $k_1$ | no início | $(a x_n - b x_n y_n,\; -c y_n + d x_n y_n)$ |
| $k_2$ | no meio | $\left(a(x_n + \frac{h}{2}k_{1x}) - b(x_n + \frac{h}{2}k_{1x})(y_n + \frac{h}{2}k_{1y}),\; -c(y_n + \frac{h}{2}k_{1y}) + d(x_n + \frac{h}{2}k_{1x})(y_n + \frac{h}{2}k_{1y})\right)$ |
| $k_3$ | no meio | $\left(a(x_n + \frac{h}{2}k_{2x}) - b(x_n + \frac{h}{2}k_{2x})(y_n + \frac{h}{2}k_{2y}),\; -c(y_n + \frac{h}{2}k_{2y}) + d(x_n + \frac{h}{2}k_{2x})(y_n + \frac{h}{2}k_{2y})\right)$ |
| $k_4$ | no fim | $\left(a(x_n + h k_{3x}) - b(x_n + h k_{3x})(y_n + h k_{3y}),\; -c(y_n + h k_{3y}) + d(x_n + h k_{3x})(y_n + h k_{3y})\right)$ |

Média ponderada (1,2,2,1) dá mais peso ao meio do intervalo.

**Nota**: $k_{ix}$ e $k_{iy}$ representam as componentes de $k_i$ para presas e predadores, respectivamente.

## 8.5. Exemplo numérico

Para $a = 2.0$, $b = 0.5$, $c = 1.0$, $d = 0.2$, $h = 0.1$, partindo de $x = 2.0$, $y = 1.0$:

**Cálculo dos k's:**

$$
\begin{aligned}
k_1 &= (2.0 \cdot 2.0 - 0.5 \cdot 2.0 \cdot 1.0,\; -1.0 \cdot 1.0 + 0.2 \cdot 2.0 \cdot 1.0) = (3.0, -0.6) \\
k_2 &= \left(\begin{array}{l}
2.0(2.0 + 0.05 \cdot 3.0) - 0.5(2.0 + 0.05 \cdot 3.0)(1.0 + 0.05 \cdot (-0.6)), \\
-1.0(1.0 + 0.05 \cdot (-0.6)) + 0.2(2.0 + 0.05 \cdot 3.0)(1.0 + 0.05 \cdot (-0.6))
\end{array}\right) \\
&= (3.24, -0.51) \\
k_3 &= \left(\begin{array}{l}
2.0(2.0 + 0.05 \cdot 3.24) - 0.5(2.0 + 0.05 \cdot 3.24)(1.0 + 0.05 \cdot (-0.51)), \\
-1.0(1.0 + 0.05 \cdot (-0.51)) + 0.2(2.0 + 0.05 \cdot 3.24)(1.0 + 0.05 \cdot (-0.51))
\end{array}\right) \\
&= (3.19, -0.55) \\
k_4 &= \left(\begin{array}{l}
2.0(2.0 + 0.1 \cdot 3.19) - 0.5(2.0 + 0.1 \cdot 3.19)(1.0 + 0.1 \cdot (-0.55)), \\
-1.0(1.0 + 0.1 \cdot (-0.55)) + 0.2(2.0 + 0.1 \cdot 3.19)(1.0 + 0.1 \cdot (-0.55))
\end{array}\right) \\
&= (3.62, -0.42)
\end{aligned}
$$

**Próximas populações:**

$$
\begin{aligned}
x_{n+1} &= 2.0 + \frac{0.1}{6} \cdot (3.0 + 2 \cdot 3.24 + 2 \cdot 3.19 + 3.62) = 2.31 \\
y_{n+1} &= 1.0 + \frac{0.1}{6} \cdot (-0.6 + 2 \cdot (-0.51) + 2 \cdot (-0.55) + (-0.42)) = 0.95
\end{aligned}
$$

## 9. Ressalvas

### 9.1 Premissas do Modelo
1. **Crescimento exponencial**: Presas crescem ilimitadamente na ausência de predadores
2. **Morte exponencial**: Predadores morrem sem alimento
3. **Interação aleatória**: Encontros entre presas e predadores são aleatórios
4. **Populações homogêneas**: Todos os indivíduos são idênticos
5. **Ambiente constante**: Sem mudanças sazonais ou ambientais
6. **Sem estrutura etária**: Todos os indivíduos têm a mesma capacidade reprodutiva e mortalidade

### 9.2 Limitações do Modelo
- Em sistemas reais, a capacidade de suporte limita o crescimento das presas
- Os predadores podem ter outras fontes de alimento
- Efeitos sazonais e ambientais não são considerados
- A taxa de predação pode não ser linear (dependência funcional)
- O modelo não considera migração ou competição intraespecífica
- Para grandes amplitudes, o modelo pode produzir oscilações irreais

## 10. Generalizações

### 10.1 Lotka-Volterra com Competição Intraespecífica
Adicionando termos de competição entre as presas:

$$
\boxed{\frac{dx}{dt} = a x - b x y - \alpha x^2}
$$

### 10.2 Lotka-Volterra com Efeito Allee
Adicionando um termo de efeito Allee (crescimento mínimo necessário):

$$
\boxed{\frac{dx}{dt} = a x \left(1 - \frac{x}{K}\right) - b x y}
$$

### 10.3 Lotka-Volterra com Resposta Funcional de Holling
Adicionando uma resposta funcional não-linear (tipo II):

$$
\boxed{\frac{dx}{dt} = a x - \frac{b x y}{1 + h x}}
$$
