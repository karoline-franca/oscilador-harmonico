## 1. Fundamentos Teóricos

### 1.1 Oscilador de Van der Pol
O oscilador de Van der Pol descreve um sistema não linear com amortecimento que depende da amplitude. A dinâmica é baseada nas seguintes premissas:

1. O sistema possui um termo de amortecimento não linear que depende da posição.
2. Para pequenas amplitudes, o amortecimento é negativo (amplifica oscilações).
3. Para grandes amplitudes, o amortecimento é positivo (amortece oscilações).

### 1.2 Equação de Van der Pol
O sistema é descrito por uma equação diferencial de segunda ordem:

$$
\boxed{\frac{d^2x}{dt^2} - \mu(1 - x^2)\frac{dx}{dt} + x = 0}
$$

Onde:
- $x$ é a **posição** ou **amplitude** da oscilação
- $\mu$ é o **parâmetro de não linearidade** que controla a força do amortecimento não linear

O termo $\mu(1 - x^2)\frac{dx}{dt}$ representa o amortecimento não linear:
- Quando $|x| < 1$, o amortecimento é negativo, o que leva à amplificação da oscilação
- Quando $|x| > 1$, o amortecimento é positivo, o que leva à redução da amplitude

## 2. Pontos de Equilíbrio

### 2.1 Pontos Críticos
Para encontrar os pontos de equilíbrio, convertemos o sistema para a forma de primeira ordem:

$$
\begin{cases}
\dot{x} = y \\[5pt]
\dot{y} = \mu(1 - x^2)y - x
\end{cases}
$$

Os pontos de equilíbrio são obtidos fazendo $\dot{x} = 0$ e $\dot{y} = 0$:

**Ponto único:**
$$
\boxed{(x^*, y^*) = (0, 0)}
$$

### 2.2 Estabilidade dos Pontos de Equilíbrio
- **Ponto (0,0)**: 
  - Para $\mu = 0$: Centro (neutramente estável) - oscilações harmônicas
  - Para $\mu > 0$: Foco instável para pequenas perturbações (as trajetórias espiralam para fora até atingirem o ciclo limite)
  - Para $\mu < 0$: Foco estável para pequenas perturbações

## 3. Solução Analítica

### 3.1 Comportamento Assintótico
O sistema de Van der Pol não possui uma constante de movimento simples, em vez disso, o sistema possui um **ciclo limite estável** para $\mu > 0$.

Para $\mu \ll 1$, a solução pode ser aproximada por:

$$
\boxed{x(t) \approx 2\cos(t) + \frac{\mu}{4}(3\sin(t) - \sin(3t)) + \mathcal{O}(\mu^2)}
$$

### 3.2 Ciclo Limite
Para $\mu \gg 1$, a oscilação se torna uma **oscilação de relaxação**, com dois regimes distintos:

1. **Regime lento**: A trajetória segue aproximadamente a curva $\dot{x} = 0$ (ou seja, $y \approx 0$)
2. **Regime rápido**: Transições abruptas entre os estados lentos

O ciclo limite pode ser encontrado numericamente, pois não há solução analítica fechada.

## 4. Parâmetros do Sistema

### 4.1 Período de Oscilação
Para $\mu \ll 1$, o período é aproximadamente:

$$
\boxed{T \approx 2\pi}
$$

Para $\mu \gg 1$, o período escala como:

$$
\boxed{T \approx \frac{3 - 2\ln(2)}{\mu} \approx \frac{1.6137}{\mu}}
$$

### 4.2 Frequência Angular
Para $\mu \ll 1$, a frequência angular é:

$$
\boxed{\omega_0 \approx 1}
$$

### 4.3 Amplitude do Ciclo Limite
A amplitude do ciclo limite é aproximadamente constante e independe do valor de $\mu$:

$$
\boxed{A \approx 2}
$$

## 5. Constante de Movimento

### 5.1 Função de Energia
Podemos definir uma função de "energia" para o sistema:

$$
E(x,y) = \frac{1}{2}y^2 + \frac{1}{2}x^2
$$

### 5.2 Variação da Energia
Ao longo de uma trajetória, a derivada total da energia é:

$$
\frac{dE}{dt} = y\dot{y} + x\dot{x} = y(\mu(1 - x^2)y - x) + xy = \mu(1 - x^2)y^2
$$

A energia do sistema **não é conservada**. A variação de energia depende do termo de amortecimento não linear:
- Quando $|x| < 1$, $\frac{dE}{dt} > 0$ (sistema ganha energia)
- Quando $|x| > 1$, $\frac{dE}{dt} < 0$ (sistema perde energia)

### 5.3 Ciclo Limite
A energia se estabiliza em um valor constante correspondente ao ciclo limite:

$$
\boxed{E_{\text{limite}} \approx 2}
$$

## 6. Sistema de Equações de Primeira Ordem

Para aplicação do **Runge-Kutta**, o sistema deve ser convertido para a forma de primeira ordem.

Definindo:
- $y_1 = x$ (posição)
- $y_2 = \dot{x}$ (velocidade)

O sistema fica:

$$
\begin{cases}
\dot{y}_1 = y_2 \\[10pt]
\dot{y}_2 = \mu(1 - y_1^2)y_2 - y_1
\end{cases}
$$

## 7. Espaço de Fases

O **espaço de fases** é definido pelos eixos $(x, \dot{x})$, onde cada ponto representa o estado instantâneo do sistema.

### 7.1 Trajetória no Espaço de Fases
Para o oscilador de Van der Pol com $\mu > 0$, as trajetórias são **curvas espirais** que convergem para o ciclo limite:

- Para condições iniciais dentro do ciclo: trajetórias espiralam para fora
- Para condições iniciais fora do ciclo: trajetórias espiralam para dentro
- No ciclo limite: trajetórias periódicas

### 7.2 Características das Órbitas
- **Forma aproximadamente circular**: Para $\mu \ll 1$
- **Forma curva com transições abruptas (oscilação de relaxação)**: Para $\mu \gg 1$
- **Centro**: Ponto de equilíbrio $(0,0)$
- **Sentido**: Anti-horário (oscilações periódicas)
- **Amplitude**: Aproximadamente 2 (ciclo limite)

### 7.3 Ciclo de Oscilação (para $\mu \gg 1$)
1. **Fase lenta (subida)**: $x$ aumenta lentamente de $-2$ para $2$
2. **Fase rápida (descida)**: $x$ cai rapidamente de $2$ para $-2$
3. **Fase lenta (descida)**: $x$ diminui lentamente de $-2$ para $2$ (na verdade, é a mesma fase com orientação oposta)
4. **Fase rápida (subida)**: $x$ sobe rapidamente de $-2$ para $2$

## 8. Método Runge-Kutta de 4ª Ordem (RK4)

Para resolver numericamente o sistema, utilizamos o método RK4:

Dado o sistema $\dot{\mathbf{y}} = \mathbf{f}(t, \mathbf{y})$, com passo $h = \Delta t$, temos como equações principais:

$$
\begin{aligned}
\mathbf{k}_1 &= \mathbf{f}(t_n, \mathbf{y}_n) \\
\mathbf{k}_2 &= \mathbf{f}\left(t_n + \frac{h}{2}, \mathbf{y}_n + \frac{h}{2}\mathbf{k}_1\right) \\
\mathbf{k}_3 &= \mathbf{f}\left(t_n + \frac{h}{2}, \mathbf{y}_n + \frac{h}{2}\mathbf{k}_2\right) \\
\mathbf{k}_4 &= \mathbf{f}\left(t_n + h, \mathbf{y}_n + h\mathbf{k}_3\right)
\end{aligned}
$$

$$
\boxed{\mathbf{y}_{n+1} = \mathbf{y}_n + \frac{h}{6}(\mathbf{k}_1 + 2\mathbf{k}_2 + 2\mathbf{k}_3 + \mathbf{k}_4)}
$$

## 8.1. O que é o RK4?

É uma fórmula que calcula a posição e velocidade no próximo instante de tempo utilizando 4 estimativas das derivadas dentro do intervalo.

## 8.2. O problema

O sistema de Van der Pol na forma de primeira ordem é:

$$
\begin{cases}
x' = y \\
y' = \mu(1 - x^2)y - x
\end{cases}
$$

## 8.3. A fórmula

Para avançar de $(x_n, y_n)$ para $(x_{n+1}, y_{n+1})$ com passo $h$:

$$
\boxed{(x_{n+1}, y_{n+1}) = (x_n, y_n) + \frac{h}{6}(k_1 + 2k_2 + 2k_3 + k_4)}
$$

## 8.4. As variáveis

| k | Onde calcula | Como calcula |
|---|--------------|--------------|
| $k_1$ | no início | $(y_n,\; \mu(1 - x_n^2)y_n - x_n)$ |
| $k_2$ | no meio | $\left(y_n + \frac{h}{2}k_{1y},\; \mu(1 - (x_n + \frac{h}{2}k_{1x})^2)(y_n + \frac{h}{2}k_{1y}) - (x_n + \frac{h}{2}k_{1x})\right)$ |
| $k_3$ | no meio | $\left(y_n + \frac{h}{2}k_{2y},\; \mu(1 - (x_n + \frac{h}{2}k_{2x})^2)(y_n + \frac{h}{2}k_{2y}) - (x_n + \frac{h}{2}k_{2x})\right)$ |
| $k_4$ | no fim | $\left(y_n + h k_{3y},\; \mu(1 - (x_n + h k_{3x})^2)(y_n + h k_{3y}) - (x_n + h k_{3x})\right)$ |

Média ponderada (1,2,2,1) dá mais peso ao meio do intervalo.

**Nota**: $k_{ix}$ e $k_{iy}$ representam as componentes de $k_i$ para posição e velocidade, respectivamente.

## 8.5. Exemplo numérico

Para $\mu = 1.0$, $h = 0.1$, partindo de $x = 0.5$, $y = 0.0$:

**Cálculo dos k's:**

$$
\begin{aligned}
k_1 &= (0.0,\; 1.0(1 - 0.5^2)\cdot 0.0 - 0.5) = (0.0, -0.5) \\
k_2 &= \left(\begin{array}{l}
0.0 + 0.05 \cdot (-0.5), \\
1.0(1 - (0.5 + 0.05 \cdot 0.0)^2)(0.0 + 0.05 \cdot (-0.5)) - (0.5 + 0.05 \cdot 0.0)
\end{array}\right) \\
&= (-0.025, -0.4875) \\
k_3 &= \left(\begin{array}{l}
0.0 + 0.05 \cdot (-0.4875), \\
1.0(1 - (0.5 + 0.05 \cdot (-0.025))^2)(0.0 + 0.05 \cdot (-0.4875)) - (0.5 + 0.05 \cdot (-0.025))
\end{array}\right) \\
&= (-0.0244, -0.4871) \\
k_4 &= \left(\begin{array}{l}
0.0 + 0.1 \cdot (-0.4871), \\
1.0(1 - (0.5 + 0.1 \cdot (-0.0244))^2)(0.0 + 0.1 \cdot (-0.4871)) - (0.5 + 0.1 \cdot (-0.0244))
\end{array}\right) \\
&= (-0.0487, -0.4747)
\end{aligned}
$$

**Próximos estados:**

$$
\begin{aligned}
x_{n+1} &= 0.5 + \frac{0.1}{6} \cdot (0.0 + 2 \cdot (-0.025) + 2 \cdot (-0.0244) + (-0.0487)) = 0.4952 \\
y_{n+1} &= 0.0 + \frac{0.1}{6} \cdot (-0.5 + 2 \cdot (-0.4875) + 2 \cdot (-0.4871) + (-0.4747)) = -0.0487
\end{aligned}
$$

## 9. Ressalvas

### 9.1 Premissas do Modelo
1. **Amortecimento não linear**: O amortecimento depende da amplitude da oscilação
2. **Sistema autônomo**: O sistema não depende explicitamente do tempo
3. **Oscilador único**: Apenas uma variável de posição e uma de velocidade
4. **Parâmetro $\mu$ constante**: O parâmetro de não linearidade é fixo
5. **Sem forçamento externo**: Não há termos de excitação externa

### 9.2 Limitações do Modelo
- Em sistemas reais, o amortecimento pode ter formas mais complexas
- O parâmetro $\mu$ pode variar com o tempo ou com a frequência
- O modelo não considera ruído ou perturbações externas
- Para $\mu$ muito grande, o sistema se torna stiff e requer métodos numéricos especiais
- O modelo pode não capturar todos os fenômenos não lineares observados em sistemas reais
- O ciclo limite é estável apenas para $\mu > 0$
