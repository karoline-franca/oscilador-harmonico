```python
resumo_ciclo = []

for (sistema, x0, v0), grupo in df.groupby(
    ['sistema_id', 'x0', 'v0']
):

    df_aux = grupo.copy()

    df_aux['posicao'] = pd.to_numeric(
        df_aux['posicao'],
        errors='coerce'
    )

    df_aux['velocidade'] = pd.to_numeric(
        df_aux['velocidade'],
        errors='coerce'
    )

    # arredondamento para evitar erro numérico
    df_aux['posicao_round'] = (
        df_aux['posicao'].round(6)
    )

    df_aux['velocidade_round'] = (
        df_aux['velocidade'].round(6)
    )

    # ordena temporalmente
    df_aux = df_aux.sort_values('tempo')

    # identifica repetição do par (x,v)
    repetido = df_aux.duplicated(
        subset=['posicao_round', 'velocidade_round'],
        keep='first'
    )

    # pega o primeiro instante em que ocorre repetição
    if repetido.any():

        idx_primeira_repeticao = repetido.idxmax()

        tempo_fechamento = (
            df_aux.loc[idx_primeira_repeticao, 'tempo']
        )

    else:

        tempo_fechamento = None

    resumo_ciclo.append({
        'sistema_id': sistema,
        'x0': x0,
        'v0': v0,
        'tempo_fechamento_ciclo': tempo_fechamento
    })

df_fechamento_ciclo = pd.DataFrame(resumo_ciclo)

df_fechamento_ciclo
```

```python
# ---------------------------------------------
# Objetivo:
# Encontrar o instante em que o sistema
# retorna aproximadamente ao estado inicial
# (x0, v0) no espaço de fases.
#
# Considera:
# - tempo discretizado com dt = 0.01
# - retorno aproximado (não exato)
# ---------------------------------------------
```

```python
dt = 0.01

resumo_ciclo = []

for (sistema, x0, v0), grupo in df.groupby(
    ['sistema_id', 'x0', 'v0']
):

    # ordena temporalmente
    df_aux = (
        grupo
        .sort_values('tempo')
        .copy()
    )

    # estado inicial
    x_inicial = df_aux.iloc[0]['posicao']
    v_inicial = df_aux.iloc[0]['velocidade']

    # remove instante inicial
    df_verifica = df_aux.iloc[1:].copy()

    # distância euclidiana no espaço de fases
    df_verifica['distancia_fase'] = np.sqrt(
        (df_verifica['posicao'] - x_inicial)**2
        +
        (df_verifica['velocidade'] - v_inicial)**2
    )

    # encontra o ponto mais próximo do estado inicial
    idx_min = df_verifica['distancia_fase'].idxmin()

    tempo_retorno = (
        df_verifica.loc[idx_min, 'tempo']
    )

    distancia_min = (
        df_verifica.loc[idx_min, 'distancia_fase']
    )

    # período teórico do OHS
    omega = df_aux.iloc[0]['frequencia_angular']

    periodo_teorico = (
        2 * np.pi / omega
    )

    # erro em relação ao período teórico
    erro_periodo = abs(
        tempo_retorno - periodo_teorico
    )

    resumo_ciclo.append({

        'sistema_id': sistema,
        'x0': x0,
        'v0': v0,

        'frequencia_angular': omega,

        'dt': dt,

        'tempo_retorno_aprox': tempo_retorno,

        'periodo_teorico': periodo_teorico,

        'erro_periodo': erro_periodo,

        'distancia_minima_fase': distancia_min

    })

df_retorno_ciclo = pd.DataFrame(resumo_ciclo)

df_retorno_ciclo
```

```python
# ---------------------------------------------
# Objetivo:
# Encontrar todos os pares (x,v) repetidos
# considerando aproximação numérica devido
# à discretização temporal.
#
# Estratégia:
# - arredonda (x,v)
# - identifica repetições do par completo
# ---------------------------------------------
```

```python
casos_repetidos = []

casas_decimais = 5

for (sistema, x0, v0), grupo in df.groupby(
    ['sistema_id', 'x0', 'v0']
):

    df_aux = grupo.copy()

    # arredondamento numérico
    df_aux['x_round'] = (
        df_aux['posicao']
        .round(casas_decimais)
    )

    df_aux['v_round'] = (
        df_aux['velocidade']
        .round(casas_decimais)
    )

    # contagem de pares (x,v)
    contagem = (
        df_aux
        .groupby(['x_round', 'v_round'])
        .size()
        .reset_index(name='qtd')
    )

    # mantém apenas pares repetidos
    repetidos = (
        contagem[
            contagem['qtd'] > 1
        ]
        .copy()
    )

    # adiciona metadados
    repetidos['sistema_id'] = sistema
    repetidos['x0'] = x0
    repetidos['v0'] = v0

    casos_repetidos.append(repetidos)

df_pares_repetidos = pd.concat(
    casos_repetidos,
    ignore_index=True
)

df_pares_repetidos = df_pares_repetidos[
    [
        'sistema_id',
        'x0',
        'v0',
        'x_round',
        'v_round',
        'qtd'
    ]
]

df_pares_repetidos
```
