Este notebook contém a solução do OHS utilizando Runge-Kutta. 
\
Além disso, gera uma base de dados com as posições e velocidades do OHS no tempo, obtendo assim o espaço de fase do sistema.

# **Descrição do Problema**



## 1. Fundamentos Teóricos

### 1.1 Lei de Hooke
A força restauradora em um sistema massa-mola é descrita pela **Lei de Hooke**:

$$
\boxed{F = -k x}
$$

Onde:
- $F$ é a **força restauradora**
- $k$ é a **constante elástica**
- $x$ é o **deslocamento** em relação à posição de equilíbrio do sistema

O sinal negativo indica que a força atua no sentido oposto ao deslocamento, caracterizando assim um movimento oscilatório.

### 1.2 Segunda Lei de Newton
Aplicando a **Segunda Lei de Newton** ao sistema:

$$
\boxed{F = m a = m \frac{d^2x}{dt^2}}
$$

Onde:
- $m$ é a **massa** do corpo
- $a = \frac{d^2x}{dt^2}$ é a **aceleração** do corpo

## 2. Equação de Movimento

Igualando as expressões anteriores:

$$
m \frac{d^2x}{dt^2} = -k x
$$

Obtemos a EDO que descreve o oscilador harmônico simples (OHS):

$$
\boxed{\frac{d^2x}{dt^2} + \frac{k}{m} x = 0}
$$

Cuja **frequência angular** é:

$$
\omega_0 = \sqrt{\frac{k}{m}}
$$

Substituindo:

$$
\boxed{\frac{d^2x}{dt^2} + \omega_0^2 x = 0}
$$

## 3. Solução Analítica

### 3.1 Solução Geral
A equação diferencial é linear e homogênea, de segunda ordem, logo, sua solução geral é:

$$
\boxed{x(t) = A \cos(\omega_0 t + \phi)}
$$

Onde:
- $A$ é a **amplitude máxima** do movimento
- $\phi$ é a **fase inicial**
- $\omega_0$ é a **frequência angular**

### 3.2 Velocidade e Aceleração
Derivando em relação ao tempo, obtemos:

**Velocidade:**
$$
v(t) = \frac{dx}{dt} = -A \omega_0 \sin(\omega_0 t + \phi)
$$

**Aceleração:**
$$
a(t) = \frac{d^2x}{dt^2} = -A \omega_0^2 \cos(\omega_0 t + \phi) = -\omega_0^2 x(t)
$$

### 3.3 Constantes de Movimento
As constantes $A$ e $\phi$ são determinadas pelas **condições iniciais**:

Para $t = 0$:
- Posição inicial: $x(0) = x_0 = A \cos(\phi)$
- Velocidade inicial: $v(0) = v_0 = -A \omega_0 \sin(\phi)$

A amplitude é dada por:
$$
\boxed{A = \sqrt{x_0^2 + \left(\frac{v_0}{\omega_0}\right)^2}}
$$

A fase inicial é:
$$
\boxed{\phi = \arctan\left(-\frac{v_0}{\omega_0 x_0}\right)}
$$

## 4. Parâmetros do Sistema

### 4.1 Período de Oscilação
O tempo necessário para completar um ciclo é:

$$
\boxed{T = \frac{2\pi}{\omega_0} = 2\pi \sqrt{\frac{m}{k}}}
$$

### 4.2 Frequência
O número de oscilações por unidade de tempo é:

$$
\boxed{f = \frac{1}{T} = \frac{\omega_0}{2\pi} = \frac{1}{2\pi} \sqrt{\frac{k}{m}}}
$$

$\omega_0 = \frac{2\pi}{T} = 2\pi f$ (a frequência independe da amplitude)

## 5. Energia do Sistema

### 5.1 Energia Cinética
$$
\boxed{E_c(t) = \frac{1}{2} m v(t)^2 = \frac{1}{2} m A^2 \omega_0^2 \sin^2(\omega_0 t + \phi)}
$$

### 5.2 Energia Potencial Elástica
$$
\boxed{E_p(t) = \frac{1}{2} k x(t)^2 = \frac{1}{2} k A^2 \cos^2(\omega_0 t + \phi)}
$$

### 5.3 Energia Mecânica Total
Seja $k = m \omega_0^2$, então:

$$
E_{total}(t) = E_c(t) + E_p(t) = \frac{1}{2} m A^2 \omega_0^2 [\sin^2(\omega_0 t + \phi) + \cos^2(\omega_0 t + \phi)]
$$

$$
\boxed{E_{T} = \frac{1}{2} m \omega_0^2 A^2 = \frac{1}{2} k A^2 = \text{const.}}
$$

Ou seja, no OHS ideal (sem amortecimento) a energia mecânica total é **conservada**.

## 6. Sistema de Equações de Primeira Ordem

Para aplicação do **Runge-Kutta**, transformamos a EDO de segunda ordem em um sistema de duas EDOs de primeira ordem:

Definindo:
- $y_1 = x$ (posição)
- $y_2 = \dot{x} = v$ (velocidade)

O sistema fica:

$$
\begin{cases}
\dot{y}_1 = y_2 \\[10pt]
\dot{y}_2 = -\omega_0^2 y_1
\end{cases}
$$

## 7. Espaço de Fases

O **espaço de fases** é definido pelos eixos $(x, v)$, onde cada ponto representa o estado instantâneo do sistema.

### 7.1 Trajetória no Espaço de Fases
Para o OHS, a trajetória é uma elipse:

$$
\boxed{\frac{x^2}{A^2} + \frac{v^2}{(\omega_0 A)^2} = 1}
$$

### 7.2 Características da Elipse
- **Semi-eixo horizontal**: $A$ (amplitude máxima)
- **Semi-eixo vertical**: $\omega_0 A$ (velocidade máxima)
- A área da elipse: $\pi A (\omega_0 A) = \pi \omega_0 A^2$ (proporcional ao quadrado da amplitude total do sistema)

## 8. Método Runge-Kutta de 4ª Ordem (RK4)

Para resolver numericamente o sistema, utilizamos o método RK4:

Dado o sistema $\dot{\mathbf{y}} = \mathbf{f}(t, \mathbf{y})$, com passo $h = \Delta t$, temos como equações principais:

$$
\begin{aligned}
\mathbf{k}_1 &= \mathbf{f}(t_n, \mathbf{y}_n) \\
\mathbf{k}_2 &= \mathbf{f}\left(t_n + \frac{h}{2}, \mathbf{y}_n + \frac{h}{2}\mathbf{k}_1\right) \\
\mathbf{k}_3 &= \mathbf{f}\left(t_n + \frac{h}{2}, \mathbf{y}_n + \frac{h}{2}\mathbf{k}_2\right) \\
\mathbf{k}_4 &= \mathbf{f}\left(t_n + h, \mathbf{y}_n + h\mathbf{k}_3\right)
\end{aligned}
$$

$$
\boxed{\mathbf{y}_{n+1} = \mathbf{y}_n + \frac{h}{6}(\mathbf{k}_1 + 2\mathbf{k}_2 + 2\mathbf{k}_3 + \mathbf{k}_4)}
$$

## 8.1. O que é o RK4?

É uma fórmula que calcula posição e velocidade no próximo instante de tempo utilizando 4 estimativas da derivada dentro do intervalo.

## 8.2. O problema

A equação $x'' + \omega_0^2 x = 0$ se torna:

$$
\begin{cases}
x' = v \\
v' = -\omega_0^2 x
\end{cases}
$$

## 8.3. A fórmula

Para avançar de $(x_n, v_n)$ para $(x_{n+1}, v_{n+1})$ com passo $h$:

$$
\boxed{(x_{n+1}, v_{n+1}) = (x_n, v_n) + \frac{h}{6}(k_1 + 2k_2 + 2k_3 + k_4)}
$$

## 8.4. As variáveis

| k | Onde calcula | Como calcula |
|---|--------------|--------------|
| $k_1$ | no início | $(v_n,\; -\omega_0^2 x_n)$ |
| $k_2$ | no meio | $\left(v_n + \frac{h}{2}k_{1v},\; -\omega_0^2\left(x_n + \frac{h}{2}k_{1x}\right)\right)$ |
| $k_3$ | no meio | $\left(v_n + \frac{h}{2}k_{2v},\; -\omega_0^2\left(x_n + \frac{h}{2}k_{2x}\right)\right)$ |
| $k_4$ | no fim | $\left(v_n + h k_{3v},\; -\omega_0^2\left(x_n + h k_{3x}\right)\right)$ |

Média ponderada (1,2,2,1) dá mais peso ao meio do intervalo

## 8.5. Exemplo numérico

Para $\omega_0 = 1$, $h = 0.1$, partindo de $x = 1$, $v = 0$:

**Cálculo dos k's:**

$$
\begin{aligned}
k_1 &= (0,\; -1 \cdot 1) = (0, -1) \\
k_2 &= \left[0 + \frac{0.1}{2} \cdot 0,\; -1 \cdot \left(1 + \frac{0.1}{2} \cdot 0\right)\right] = (0, -1) \\
k_3 &= \left[0 + \frac{0.1}{2} \cdot 0,\; -1 \cdot \left(1 + \frac{0.1}{2} \cdot 0\right)\right] = (0, -1) \\
k_4 &= \left[0 + 0.1 \cdot 0,\; -1 \cdot \left(1 + 0.1 \cdot 0\right)\right] = (0, -1)
\end{aligned}
$$

**Próximas posição e velocidade:**

$$
\begin{aligned}
x_{n+1} &= 1 + \frac{0.1}{6} \cdot (0 + 2 \cdot 0 + 2 \cdot 0 + 0) = 1 \\
v_{n+1} &= 0 + \frac{0.1}{6} \cdot [-1 + 2 \cdot (-1) + 2 \cdot (-1) + (-1)] = -0.1
\end{aligned}
$$

## 9. Ressalvas

### 9.1 Premissas do Modelo
1. **Mola ideal**: massa desprezível, obedece à Lei de Hooke
2. **Sem amortecimento**: não há forças dissipativas (atrito, resistência do ar)
3. **Movimento unidimensional**: restrito ao eixo x
4. **Ponto material**: massa concentrada em um ponto

### 9.2 Limitações do Modelo
- Para grandes amplitudes, a Lei de Hooke pode não ser válida (regime não linear)
- Na prática, sempre há alguma força dissipativa
- A massa da mola pode ser significativa para sistemas muito leves

## 10. Generalizações

### 10.1 Oscilador Harmônico Amortecido
Adicionando uma força de amortecimento proporcional à velocidade ($F_{amort} = -b v$):

$$
\boxed{m\frac{d^2x}{dt^2} + b\frac{dx}{dt} + kx = 0}
$$

### 10.2 Oscilador Harmônico Forçado
Adicionando uma força externa periódica ($F_{ext} = F_0 \cos(\omega t)$):

$$
\boxed{m\frac{d^2x}{dt^2} + b\frac{dx}{dt} + kx = F_0 \cos(\omega t)}
$$

## Principais Equações

| Grandeza | Equação |
|----------|---------|
| Força restauradora | $F = -kx$ |
| Equação do movimento | $\frac{d^2x}{dt^2} + \omega_0^2 x = 0$ |
| Solução | $x(t) = A\cos(\omega_0 t + \phi)$ |
| Período | $T = 2\pi\sqrt{\frac{m}{k}}$ |
| Frequência linear | $f = \frac{1}{2\pi}\sqrt{\frac{k}{m}}$ |
| Frequência angular | $\omega_0 = \sqrt{\frac{k}{m}}$ |
| Amplitude máxima | $A = \sqrt{x_0^2 + \left(\frac{v_0}{\omega_0}\right)^2}$ |
| Energia total | $E = \frac{1}{2}kA^2$ |
| Espaço de fases | $\frac{x^2}{A^2} + \frac{v^2}{(\omega_0 A)^2} = 1$ |

## Equação a ser Resolvida com RK4

$$
\boxed{
\begin{cases}
\dfrac{dx}{dt} = v \\[10pt]
\dfrac{dv}{dt} = -\dfrac{k}{m}x - \dfrac{b}{m}v
\end{cases}
}
$$

# **Simulação**

```python
class OsciladorHarmonicoPyTorch:
    """
    Classe para resolver a equação do oscilador harmônico simples usando PyTorch e Runge-Kutta
    """
    
    def __init__(self, massa, constante_elastica, amortecimento=0.0, device='cpu'):
        """
        Parâmetros do oscilador
        
        Args:
            massa (float): massa do sistema (kg)
            constante_elastica (float): força restauradora (N/m)
            amortecimento (float): coeficiente de amortecimento (kg/s)
            device (str): dispositivo para computação ('cpu' ou 'cuda')
        """
        self.m = torch.tensor(massa, dtype=torch.float32, device=device)
        self.k = torch.tensor(constante_elastica, dtype=torch.float32, device=device)
        self.b = torch.tensor(amortecimento, dtype=torch.float32, device=device)
        self.device = device
        self.frequencia_natural_hz = torch.sqrt(self.k / self.m).item() / (2 * np.pi)
        self.frequencia_angular = torch.sqrt(self.k / self.m).item()
        
    def equacoes_movimento(self, estados):
        """
        Define as equações do movimento para múltiplas condições iniciais
        """
        x = estados[:, 0]
        v = estados[:, 1]
        
        dxdt = v
        dvdt = -(self.k / self.m) * x - (self.b / self.m) * v
        
        return torch.stack([dxdt, dvdt], dim=1)
    
    def runge_kutta_4(self, estados, dt):
        """
        Método Runge-Kutta de 4ª ordem para múltiplas condições iniciais
        """
        k1 = self.equacoes_movimento(estados)
        k2 = self.equacoes_movimento(estados + 0.5 * dt * k1)
        k3 = self.equacoes_movimento(estados + 0.5 * dt * k2)
        k4 = self.equacoes_movimento(estados + dt * k3)
        
        estados_novos = estados + (dt / 6.0) * (k1 + 2*k2 + 2*k3 + k4)
        
        return estados_novos
        
    def resolver_multi_condicoes(self, condicoes_iniciais, t_final, dt):
        """
        Resolve a EDO para múltiplas condições iniciais simultaneamente
        """
        n_condicoes = condicoes_iniciais.shape[0]
        n_passos = int(t_final / dt) + 1
        
        tempos = torch.linspace(0, t_final, n_passos, device=self.device)
        posicoes = torch.zeros((n_passos, n_condicoes), device=self.device)
        velocidades = torch.zeros((n_passos, n_condicoes), device=self.device)
        
        estados = condicoes_iniciais.clone()
        posicoes[0] = estados[:, 0]
        velocidades[0] = estados[:, 1]
        
        for i in range(1, n_passos):
            estados = self.runge_kutta_4(estados, dt)
            posicoes[i] = estados[:, 0]
            velocidades[i] = estados[:, 1]
        
        energia_cinetica = 0.5 * self.m * velocidades**2
        energia_potencial = 0.5 * self.k * posicoes**2
        energia_mecanica = energia_cinetica + energia_potencial
        amplitudes_max = torch.max(torch.abs(posicoes), dim=0)[0].cpu().numpy()
        
        return {
            'tempo': tempos.cpu().numpy(),
            'posicao': posicoes.cpu().numpy(),
            'velocidade': velocidades.cpu().numpy(),
            'energia_cinetica': energia_cinetica.cpu().numpy(),
            'energia_potencial': energia_potencial.cpu().numpy(),
            'energia_mecanica': energia_mecanica.cpu().numpy(),
            'amplitudes': amplitudes_max,
            'condicoes_iniciais': condicoes_iniciais.cpu().numpy(),
            'massa': self.m.item(),
            'constante_elastica': self.k.item(),
            'frequencia_natural_hz': self.frequencia_natural_hz,
            'frequencia_angular_rads': self.frequencia_angular
        }
    
    def formatar_numero_pt_br(self, numero):
        """
        Formata número no padrão brasileiro
        """
        if isinstance(numero, (int, float)):
            if abs(numero) < 1000:
                return f"{numero:.3f}".replace('.', ',')
            else:
                partes = f"{numero:.3f}".split('.')
                parte_inteira = '{:,.0f}'.format(int(partes[0])).replace(',', '.')
                return f"{parte_inteira},{partes[1]}"
        return str(numero)
    
    def gerar_base_dados(self, condicoes_iniciais, t_final, dt, sistema_id, descricao_sistema):
        """
        Gera uma base de dados completa com as soluções no formato brasileiro
        
        Args:
            condicoes_iniciais: tensor com condições iniciais
            t_final: tempo final
            dt: passo temporal
            sistema_id: identificador único do sistema
            descricao_sistema: descrição textual do sistema
        """
        solucao = self.resolver_multi_condicoes(condicoes_iniciais, t_final, dt)
        
        dados = []
        n_condicoes = condicoes_iniciais.shape[0]
        n_passos = len(solucao['tempo'])
        
        for i in range(n_condicoes):
            for j in range(n_passos):
                dados.append({
                    'sistema_id': sistema_id,
                    'simulacao_id': i,
                    'massa_kg': self.formatar_numero_pt_br(solucao['massa']),
                    'constante_elastica_nm': self.formatar_numero_pt_br(solucao['constante_elastica']),
                    'frequencia_natural_hz': self.formatar_numero_pt_br(solucao['frequencia_natural_hz']),
                    'frequencia_angular_rads': self.formatar_numero_pt_br(solucao['frequencia_angular_rads']),
                    'x0': self.formatar_numero_pt_br(solucao['condicoes_iniciais'][i, 0]),
                    'v0': self.formatar_numero_pt_br(solucao['condicoes_iniciais'][i, 1]),
                    'amplitude_max': self.formatar_numero_pt_br(solucao['amplitudes'][i]),
                    'energia_cinetica': self.formatar_numero_pt_br(solucao['energia_cinetica'][j, i]),
                    'energia_potencial': self.formatar_numero_pt_br(solucao['energia_potencial'][j, i]),
                    'energia_mecanica': self.formatar_numero_pt_br(solucao['energia_mecanica'][j, i]),
                    'tempo': self.formatar_numero_pt_br(solucao['tempo'][j]),
                    'posicao': self.formatar_numero_pt_br(solucao['posicao'][j, i]),
                    'velocidade': self.formatar_numero_pt_br(solucao['velocidade'][j, i]),
                })
        
        return pd.DataFrame(dados), solucao
```

# **Soluções Representativas**

```python
def selecionar_trajetorias_representativas(solucao, n_trajetorias=20):
    """
    Seleciona trajetórias representativas baseado nas características físicas relevantes
    """
    n_total = solucao['amplitudes'].shape[0]
    amplitudes = solucao['amplitudes']
    cond_iniciais = solucao['condicoes_iniciais']
    
    # normaliza amplitudes para seleção
    amplitudes_norm = amplitudes / amplitudes.max() if amplitudes.max() > 0 else amplitudes
    
    # cria categorias baseadas na amplitude
    categorias = []
    for i in range(n_total):
        # classifica por amplitude (pequena, média, grande)
        if amplitudes_norm[i] < 0.33:
            cat_amp = 0
        elif amplitudes_norm[i] < 0.66:
            cat_amp = 1
        else:
            cat_amp = 2
        
        # classificar por fase (razão x0/v0)
        x0, v0 = cond_iniciais[i]
        if abs(x0) > abs(v0):
            cat_fase = 0  # dominância de posição
        elif abs(v0) > abs(x0):
            cat_fase = 1  # dominância de velocidade
        else:
            cat_fase = 2  # equilibrado
        
        categorias.append((cat_amp, cat_fase))
    
    # seleciona trajetórias representativas de cada categoria
    indices_selecionados = []
    categorias_unicas = list(set(categorias))
    
    # garante que peguemos de todas as categorias
    for cat in categorias_unicas:
        indices_cat = [i for i, c in enumerate(categorias) if c == cat]
        n_por_cat = max(1, int(np.ceil(n_trajetorias / len(categorias_unicas))))
        indices_selecionados.extend(indices_cat[:n_por_cat])
    
    # se ainda não temos o suficiente, completa com as de maior energia
    if len(indices_selecionados) < n_trajetorias:
        energias_totais = solucao['energia_mecanica'][-1]
        indices_restantes = [i for i in range(n_total) if i not in indices_selecionados]
        indices_restantes.sort(key=lambda i: energias_totais[i], reverse=True)
        indices_selecionados.extend(indices_restantes[:n_trajetorias - len(indices_selecionados)])
    
    return indices_selecionados[:n_trajetorias]
```

# **Gráficos**

```python
import numpy as np
import pandas as pd
import torch
import locale
import plotly.graph_objects as go
from plotly.subplots import make_subplots
# define o formato padrão brasileiro numérico dos valores da base
try:
    locale.setlocale(locale.LC_NUMERIC, 'pt_BR.UTF-8')
except:
    try:
        locale.setlocale(locale.LC_NUMERIC, 'Portuguese_Brazil.1252')
    except:
        print("Formato pt-br não disponível, usando formato padrão")
        locale.setlocale(locale.LC_NUMERIC, 'C')
class OsciladorHarmonicoPyTorch:
    """
    Classe para resolver a equação do oscilador harmônico simples usando PyTorch e Runge-Kutta
    """
    
    def __init__(self, frequencias_angulares, amortecimento=0.0, device='cpu'):
        """
        Parâmetros do oscilador para múltiplos sistemas simultâneos
        
        Args:
            frequencias_angulares (list ou tensor): frequências angulares de cada sistema ω = √(k/m) (rad/s)
            amortecimento (float): coeficiente de amortecimento (kg/s)
            device (str): dispositivo para computação ('cpu' ou 'cuda')
        """
        # Converte para tensor (n_sistemas,)
        if isinstance(frequencias_angulares, list):
            self.frequencias_angulares = torch.tensor(frequencias_angulares, dtype=torch.float32, device=device)
        else:
            self.frequencias_angulares = frequencias_angulares.clone().detach().to(device)
        
        self.amortecimento = amortecimento
        self.device = device
        self.n_sistemas = len(self.frequencias_angulares)
        
        # Converte amortecimento para tensor (expandido para todos os sistemas)
        self.b = torch.tensor(amortecimento, dtype=torch.float32, device=device).expand(self.n_sistemas)
        
        # frequência natural (em Hz) e período para cada sistema
        self.frequencias_naturais_hz = self.frequencias_angulares / (2 * np.pi)
        self.periodos = 1.0 / self.frequencias_naturais_hz
        
    def equacoes_movimento(self, estados):
        """
        Define as equações do movimento para múltiplas condições iniciais e múltiplos sistemas
        
        estados: tensor de forma (n_condicoes, n_sistemas, 2)
        retorna: tensor de forma (n_condicoes, n_sistemas, 2)
        """
        # estados[:, :, 0] = posições (n_condicoes, n_sistemas)
        # estados[:, :, 1] = velocidades (n_condicoes, n_sistemas)
        
        x = estados[:, :, 0]
        v = estados[:, :, 1]
        
        dxdt = v
        
        # Expande omega^2 para broadcast com as condições iniciais
        # omega2 tem forma (n_sistemas,) e precisa ser expandido para (n_condicoes, n_sistemas)
        omega2 = self.frequencias_angulares ** 2
        omega2_expanded = omega2.unsqueeze(0).expand(x.shape[0], -1)
        b_expanded = self.b.unsqueeze(0).expand(x.shape[0], -1)
        
        dvdt = -omega2_expanded * x - b_expanded * v
        
        return torch.stack([dxdt, dvdt], dim=2)
    
    def runge_kutta_4(self, estados, dt):
        """
        Método Runge-Kutta de 4ª ordem para múltiplas condições iniciais e múltiplos sistemas
        """
        k1 = self.equacoes_movimento(estados)
        k2 = self.equacoes_movimento(estados + 0.5 * dt * k1)
        k3 = self.equacoes_movimento(estados + 0.5 * dt * k2)
        k4 = self.equacoes_movimento(estados + dt * k3)
        
        estados_novos = estados + (dt / 6.0) * (k1 + 2*k2 + 2*k3 + k4)
        
        return estados_novos
        
    def resolver_multi_condicoes_sistemas(self, condicoes_iniciais, t_final, dt):
        """
        Resolve a EDO para múltiplas condições iniciais e múltiplos sistemas simultaneamente
        
        condicoes_iniciais: tensor de forma (n_condicoes, 2) com [x0, v0]
        t_final: tempo final (pode ser escalar ou tensor com tamanho n_sistemas)
        dt: passo temporal
        """
        n_condicoes = condicoes_iniciais.shape[0]
        n_passos = int(t_final / dt) + 1
        
        tempos = torch.linspace(0, t_final, n_passos, device=self.device)
        
        # Inicializa tensores 3D: (n_passos, n_condicoes, n_sistemas)
        posicoes = torch.zeros((n_passos, n_condicoes, self.n_sistemas), device=self.device)
        velocidades = torch.zeros((n_passos, n_condicoes, self.n_sistemas), device=self.device)
        
        # Expande condições iniciais para todos os sistemas
        # cond_iniciais_expand: (n_condicoes, n_sistemas, 2)
        cond_iniciais_expand = condicoes_iniciais.unsqueeze(1).expand(-1, self.n_sistemas, -1)
        
        estados = cond_iniciais_expand.clone()
        posicoes[0] = estados[:, :, 0]
        velocidades[0] = estados[:, :, 1]
        
        for i in range(1, n_passos):
            estados = self.runge_kutta_4(estados, dt)
            posicoes[i] = estados[:, :, 0]
            velocidades[i] = estados[:, :, 1]
        
        # calcula energias (normalizadas por massa) para cada sistema
        omega2 = self.frequencias_angulares ** 2
        omega2_expanded = omega2.unsqueeze(0).unsqueeze(0).expand(n_passos, n_condicoes, -1)
        
        energia_cinetica_especifica = 0.5 * velocidades**2
        energia_potencial_especifica = 0.5 * omega2_expanded * posicoes**2
        energia_mecanica_especifica = energia_cinetica_especifica + energia_potencial_especifica
        
        # calcula amplitudes máximas para cada condição e sistema
        amplitudes_max = torch.max(torch.abs(posicoes), dim=0)[0].cpu().numpy()
        
        return {
            'tempo': tempos.cpu().numpy(),
            'posicao': posicoes.cpu().numpy(),  # (n_passos, n_condicoes, n_sistemas)
            'velocidade': velocidades.cpu().numpy(),
            'energia_cinetica_especifica': energia_cinetica_especifica.cpu().numpy(),
            'energia_potencial_especifica': energia_potencial_especifica.cpu().numpy(),
            'energia_mecanica_especifica': energia_mecanica_especifica.cpu().numpy(),
            'amplitudes': amplitudes_max,
            'condicoes_iniciais': condicoes_iniciais.cpu().numpy(),
            'frequencias_angulares_rads': self.frequencias_angulares.cpu().numpy(),
            'frequencias_naturais_hz': self.frequencias_naturais_hz.cpu().numpy(),
            'periodos': self.periodos.cpu().numpy(),
            'n_condicoes': n_condicoes,
            'n_sistemas': self.n_sistemas
        }
    
    def formatar_numero_pt_br(self, numero):
        """
        Formata número no padrão brasileiro
        """
        if isinstance(numero, (int, float)):
            if abs(numero) < 1000:
                return f"{numero:.3f}".replace('.', ',')
            else:
                partes = f"{numero:.3f}".split('.')
                parte_inteira = '{:,.0f}'.format(int(partes[0])).replace(',', '.')
                return f"{parte_inteira},{partes[1]}"
        return str(numero)
    
    def gerar_base_dados(self, condicoes_iniciais, t_final, dt, sistemas_descricao):
        """
        Gera uma base de dados completa com as soluções no formato brasileiro
        
        Args:
            condicoes_iniciais: tensor com condições iniciais (n_condicoes, 2)
            t_final: tempo final
            dt: passo temporal
            sistemas_descricao: lista com descrições dos sistemas
        """
        solucao = self.resolver_multi_condicoes_sistemas(condicoes_iniciais, t_final, dt)
        
        dados = []
        n_condicoes = solucao['n_condicoes']
        n_sistemas = solucao['n_sistemas']
        n_passos = len(solucao['tempo'])
        
        for i_sistema in range(n_sistemas):
            for i_cond in range(n_condicoes):
                for j in range(n_passos):
                    dados.append({
                        'sistema_id': i_sistema,
                        'descricao_sistema': sistemas_descricao[i_sistema],
                        'simulacao_id': i_cond,
                        'frequencia_angular_rads': self.formatar_numero_pt_br(solucao['frequencias_angulares_rads'][i_sistema]),
                        'frequencia_natural_hz': self.formatar_numero_pt_br(solucao['frequencias_naturais_hz'][i_sistema]),
                        'periodo_s': self.formatar_numero_pt_br(solucao['periodos'][i_sistema]),
                        'x0': self.formatar_numero_pt_br(solucao['condicoes_iniciais'][i_cond, 0]),
                        'v0': self.formatar_numero_pt_br(solucao['condicoes_iniciais'][i_cond, 1]),
                        'amplitude_max': self.formatar_numero_pt_br(solucao['amplitudes'][i_cond, i_sistema]),
                        'energia_cinetica_especifica': self.formatar_numero_pt_br(solucao['energia_cinetica_especifica'][j, i_cond, i_sistema]),
                        'energia_potencial_especifica': self.formatar_numero_pt_br(solucao['energia_potencial_especifica'][j, i_cond, i_sistema]),
                        'energia_mecanica_especifica': self.formatar_numero_pt_br(solucao['energia_mecanica_especifica'][j, i_cond, i_sistema]),
                        'tempo': self.formatar_numero_pt_br(solucao['tempo'][j]),
                        'posicao': self.formatar_numero_pt_br(solucao['posicao'][j, i_cond, i_sistema]),
                        'velocidade': self.formatar_numero_pt_br(solucao['velocidade'][j, i_cond, i_sistema]),
                    })
        
        return pd.DataFrame(dados), solucao

def selecionar_trajetorias_representativas(solucao, n_trajetorias=20):
    """
    Seleciona trajetórias representativas baseado nas características físicas relevantes
    """
    n_condicoes = solucao['amplitudes'].shape[0]
    amplitudes = solucao['amplitudes']
    cond_iniciais = solucao['condicoes_iniciais']
    
    # normaliza amplitudes para seleção
    amplitudes_norm = amplitudes / amplitudes.max() if amplitudes.max() > 0 else amplitudes
    
    # cria categorias baseadas na amplitude
    categorias = []
    for i in range(n_condicoes):
        # classifica por amplitude (pequena, média, grande)
        if amplitudes_norm[i] < 0.33:
            cat_amp = 0
        elif amplitudes_norm[i] < 0.66:
            cat_amp = 1
        else:
            cat_amp = 2
        
        # classificar por fase (razão x0/v0)
        x0, v0 = cond_iniciais[i]
        if abs(x0) > abs(v0):
            cat_fase = 0  # dominância de posição
        elif abs(v0) > abs(x0):
            cat_fase = 1  # dominância de velocidade
        else:
            cat_fase = 2  # equilibrado
        
        categorias.append((cat_amp, cat_fase))
    
    # seleciona trajetórias representativas de cada categoria
    indices_selecionados = []
    categorias_unicas = list(set(categorias))
    
    # garante que peguemos de todas as categorias
    for cat in categorias_unicas:
        indices_cat = [i for i, c in enumerate(categorias) if c == cat]
        n_por_cat = max(1, int(np.ceil(n_trajetorias / len(categorias_unicas))))
        indices_selecionados.extend(indices_cat[:n_por_cat])
    
    # se ainda não temos o suficiente, completa com as de maior energia
    if len(indices_selecionados) < n_trajetorias:
        energias_totais = solucao['energia_mecanica_especifica'][-1]
        indices_restantes = [i for i in range(n_condicoes) if i not in indices_selecionados]
        # Para cada índice, pega a energia média entre os sistemas
        energias_medias = np.mean(energias_totais, axis=1)
        indices_restantes.sort(key=lambda i: energias_medias[i], reverse=True)
        indices_selecionados.extend(indices_restantes[:n_trajetorias - len(indices_selecionados)])
    
    return indices_selecionados[:n_trajetorias]

def criar_visualizacao_2d_unica(solucao, sistemas_descricao, indices_selecionados):
    """
    Cria um único gráfico 2D com todas as trajetórias de todos os sistemas no espaço de fases
    """
    fig = go.Figure()
    
    # Paleta de cores para diferentes sistemas
    cores = [
        'blue', 'red', 'green', 'purple', 'orange', 'brown',
        'pink', 'gray', 'olive', 'cyan', 'magenta', 'yellow'
    ]
    
    n_sistemas = solucao['n_sistemas']
    n_condicoes = len(indices_selecionados)
    
    for i_sistema in range(n_sistemas):
        cor = cores[i_sistema % len(cores)]
        freq_hz = solucao['frequencias_naturais_hz'][i_sistema]
        omega = solucao['frequencias_angulares_rads'][i_sistema]
        
        for i, traj_idx in enumerate(indices_selecionados):
            x0, v0 = solucao['condicoes_iniciais'][traj_idx]
            amplitude = solucao['amplitudes'][traj_idx, i_sistema]
            
            # Cria nome legível para a legenda
            if i == 0:  # Mostra apenas um item por sistema na legenda
                nome = f"Sistema {i_sistema}: ω={omega:.2f} rad/s (f={freq_hz:.2f} Hz)"
                show_legend = True
            else:
                nome = f"Traj_{i_sistema}_{traj_idx}"
                show_legend = False
            
            fig.add_trace(go.Scatter(
                x=solucao['posicao'][:, traj_idx, i_sistema],
                y=solucao['velocidade'][:, traj_idx, i_sistema],
                mode='lines',
                line=dict(color=cor, width=1.5),
                name=nome,
                legendgroup=f'sistema_{i_sistema}',
                showlegend=show_legend,
                hovertemplate=(
                    f"<b>Sistema: {sistemas_descricao[i_sistema]}</b><br>" +
                    f"ω = {omega:.3f} rad/s<br>" +
                    f"f = {freq_hz:.3f} Hz<br>" +
                    f"x₀ = {x0:.2f} m, v₀ = {v0:.2f} m/s<br>" +
                    f"A = {amplitude:.3f} m<br>" +
                    f"Posição: %{{x:.3f}} m<br>" +
                    f"Velocidade: %{{y:.3f}} m/s<br>" +
                    f"<extra></extra>"
                )
            ))
    
    fig.update_layout(
        title=(
            "Espaço de Fases 2D - Todos os Sistemas<br>" +
            "<sup>Cada cor representa um sistema diferente com suas respectivas condições iniciais</sup>"
        ),
        xaxis_title="Posição (m)",
        yaxis_title="Velocidade (m/s)",
        width=1000,
        height=800,
        legend=dict(
            title="Sistemas",
            yanchor="top",
            y=0.99,
            xanchor="left",
            x=1.02,
            bgcolor="rgba(255, 255, 255, 0.9)",
            bordercolor="Black",
            borderwidth=1,
            font=dict(size=10)
        ),
        hoverlabel=dict(
            bgcolor="white",
            font_size=12,
            font_family="Arial"
        ),
        plot_bgcolor='white',
        xaxis=dict(
            showgrid=True,
            gridcolor='lightgray',
            zeroline=True,
            zerolinecolor='black'
        ),
        yaxis=dict(
            showgrid=True,
            gridcolor='lightgray',
            zeroline=True,
            zerolinecolor='black'
        )
    )
    
    return fig

def main():
    """
    Função principal
    """
    device = 'cuda' if torch.cuda.is_available() else 'cpu'
    print(f"Usando dispositivo: {device}")
    
    # ============================================================
    # DEFINIÇÃO DE SISTEMAS TÍPICOS UTILIZANDO FREQUÊNCIA ANGULAR
    # ============================================================

    sistemas_tipicos = [
        {"id": 0, "omega": 0.5, "descricao": "Sistema Lento - Baixa frequência angular"},
        {"id": 1, "omega": 1.0, "descricao": "Sistema Médio-Baixo"},
        {"id": 2, "omega": 2.0, "descricao": "Sistema de Referência"},
        {"id": 3, "omega": 3.0, "descricao": "Sistema Médio-Alto"},
        {"id": 4, "omega": 5.0, "descricao": "Sistema Rápido"},
        {"id": 5, "omega": 8.0, "descricao": "Sistema Muito Rápido - Alta frequência angular"},
    ]
    
    # Extrai as frequências angulares e descrições
    frequencias_angulares = [sistema["omega"] for sistema in sistemas_tipicos]
    sistemas_descricao = [sistema["descricao"] for sistema in sistemas_tipicos]

    print("\n=== SISTEMAS TÍPICOS DEFINIDOS (FREQUÊNCIA ANGULAR) ===")
    for sistema in sistemas_tipicos:
        freq_hz = sistema["omega"] / (2 * np.pi)
        periodo = 1.0 / freq_hz
        print(f"  ID {sistema['id']}: {sistema['descricao']}: ω = {sistema['omega']:.4f} rad/s, f = {freq_hz:.4f} Hz, T = {periodo:.3f} s")
    
    # ============================================================
    # PARÂMETROS DA SIMULAÇÃO
    # ============================================================

    dt = 0.01
    num_periodos = 3  # número de períodos a serem simulados
    
    print("\n=== CRIANDO CONDIÇÕES INICIAIS ===")
    
    # malha de condições iniciais: 10 pontos em cada direção (10x10 = 100 simulações)
    x0_range = torch.linspace(-5.0, 5.0, 10, device=device)
    v0_range = torch.linspace(-2.5, 2.5, 10, device=device)
    
    x0_grid, v0_grid = torch.meshgrid(x0_range, v0_range, indexing='ij')
    condicoes_iniciais = torch.stack([x0_grid.flatten(), v0_grid.flatten()], dim=1)
    
    n_simulacoes_por_sistema = condicoes_iniciais.shape[0]
    n_sistemas = len(sistemas_tipicos)
    print(f"Malha de simulação: {len(x0_range)} x {len(v0_range)} = {n_simulacoes_por_sistema} simulações por sistema")
    print(f"Total de simulações simultâneas: {n_simulacoes_por_sistema} x {n_sistemas} = {n_simulacoes_por_sistema * n_sistemas}")
    
    # ============================================================
    # CRIA OSCILADOR PARA TODOS OS SISTEMAS SIMULTANEAMENTE
    # ============================================================
    
    print("\n=== CRIANDO OSCILADOR PARA MÚLTIPLOS SISTEMAS ===")
    
    # Cria um único oscilador que processa todos os sistemas
    oscilador = OsciladorHarmonicoPyTorch(
        frequencias_angulares=frequencias_angulares,
        device=device
    )
    
    # Calcula o tempo final baseado no maior período entre os sistemas
    # Isso garante que todos os sistemas completem o mesmo número de períodos
    max_periodo = max(oscilador.periodos.cpu().numpy())
    t_final = num_periodos * max_periodo
    
    print(f"Maior período entre os sistemas: {max_periodo:.3f} s")
    print(f"Simulando {num_periodos} períodos completos: t_final = {t_final:.3f} s")
    print(f"Número de sistemas processados simultaneamente: {oscilador.n_sistemas}")
    print(f"Número de condições iniciais por sistema: {n_simulacoes_por_sistema}")
    
    # ============================================================
    # GERANDO BASE DE DADOS ÚNICA PARA TODOS OS SISTEMAS
    # ============================================================
    
    print("\n=== GERANDO BASE DE DADOS PARA TODOS OS SISTEMAS SIMULTANEAMENTE ===")
    
    # Gera base de dados para todos os sistemas de uma única vez
    df, solucao = oscilador.gerar_base_dados(
        condicoes_iniciais, t_final, dt, 
        sistemas_descricao
    )
    
    print(f"Registros gerados: {len(df):,}".replace(',', '.'))
    
    # ============================================================
    # SALVA BASE DE DADOS
    # ============================================================
    
    print("\n=== SALVANDO BASE DE DADOS ===")
    df.to_csv('base_ohs_completa.csv', index=False, sep=';', encoding='utf-8')
    print(f"Base consolidada salva em 'base_ohs_completa.csv'")
    print(f"Total de registros na base consolidada: {len(df):,}".replace(',', '.'))
    print(f"Total de simulações: {n_simulacoes_por_sistema * n_sistemas}")
    print(f"Sistemas incluídos: {df['sistema_id'].nunique()}")
    
    print("\n=== PRIMEIRAS 10 LINHAS DA BASE CONSOLIDADA ===")
    print(df.head(10))
        
    print("\n=== DISTRIBUIÇÃO DOS SISTEMAS NA BASE ===")
    distribuicao = df.groupby(['sistema_id', 'descricao_sistema']).size().reset_index(name='registros')
    for _, row in distribuicao.iterrows():
        print(f"  ID {int(row['sistema_id'])} - {row['descricao_sistema']}: {row['registros']:,} registros".replace(',', '.'))
    
    # ============================================================
    # CRIA VISUALIZAÇÕES
    # ============================================================
    
    print("\n=== CRIANDO VISUALIZAÇÕES ===")
    
    # Seleciona trajetórias representativas (comum para todos os sistemas)
    indices_selecionados = selecionar_trajetorias_representativas(solucao, n_trajetorias=12)
    
    # Cria gráfico 2D único com todas as trajetórias de todos os sistemas
    fig2d_unica = criar_visualizacao_2d_unica(solucao, sistemas_descricao, indices_selecionados)
    fig2d_unica.show()
    
    # ============================================================
    # BASE CONSOLIDADA
    # ============================================================
    
    print("\n=== BASE CONSOLIDADA ===")
    print(f"Total de registros: {len(df):,}".replace(',', '.'))
    print(f"Total de sistemas simulados: {df['sistema_id'].nunique()}")
    print(f"Total de simulações por sistema: {df['simulacao_id'].nunique()}")
    
    return df, solucao, sistemas_tipicos

if __name__ == "__main__":
    df, solucao, sistemas_tipicos = main()
```
