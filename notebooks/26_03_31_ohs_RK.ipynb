Este notebook contém a solução do OHS utilizando Runge-Kutta. 
\
Além disso, é gerada uma base de dados com as posições e velocidades do OHS no tempo, obtendo assim o espaço de fase do sistema.

# **Descrição do Problema**



## 1. Fundamentos Teóricos

### 1.1 Lei de Hooke
A força restauradora em um sistema massa-mola é descrita pela **Lei de Hooke**:

$$
\boxed{F = -k x}
$$

Onde:
- $F$ é a força restauradora
- $k$ é a constante elástica da mola
- $x$ é o deslocamento em relação à posição de equilíbrio do sistema

O sinal negativo indica que a força atua no sentido oposto ao deslocamento, caracterizando assim um movimento oscilatório.

### 1.2 Segunda Lei de Newton
Aplicando a **Segunda Lei de Newton** ao sistema:

$$
\boxed{F = m a = m \frac{d^2x}{dt^2}}
$$

Onde:
- $m$ é a massa do corpo
- $a = \frac{d^2x}{dt^2}$ é a aceleração do corpo

## 2. Equação de Movimento

Igualando as expressões anteriores:

$$
m \frac{d^2x}{dt^2} = -k x
$$

Obtemos a EDO que descreve o oscilador harmônico simples (OHS):

$$
\boxed{\frac{d^2x}{dt^2} + \frac{k}{m} x = 0}
$$

Cuja **frequência angular natural** é:

$$
\omega_0 = \sqrt{\frac{k}{m}}
$$

Substituindo:

$$
\boxed{\frac{d^2x}{dt^2} + \omega_0^2 x = 0}
$$

## 3. Solução Analítica

### 3.1 Solução Geral
A equação diferencial é linear e homogênea de segunda ordem e sua solução geral é:

$$
\boxed{x(t) = A \cos(\omega_0 t + \phi)}
$$

Onde:
- $A$ é a **amplitude** máxima do movimento
- $\phi$ é a **fase inicial**
- $\omega_0$ é a **frequência angular natural**

### 3.2 Velocidade e Aceleração
Derivando em relação ao tempo, obtemos:

**Velocidade:**
$$
v(t) = \frac{dx}{dt} = -A \omega_0 \sin(\omega_0 t + \phi)
$$

**Aceleração:**
$$
a(t) = \frac{d^2x}{dt^2} = -A \omega_0^2 \cos(\omega_0 t + \phi) = -\omega_0^2 x(t)
$$

### 3.3 Constantes de Movimento
As constantes $A$ e $\phi$ são determinadas pelas **condições iniciais**:

Para $t = 0$:
- Posição inicial: $x(0) = x_0 = A \cos(\phi)$
- Velocidade inicial: $v(0) = v_0 = -A \omega_0 \sin(\phi)$

A amplitude é dada por:
$$
\boxed{A = \sqrt{x_0^2 + \left(\frac{v_0}{\omega_0}\right)^2}}
$$

A fase inicial é:
$$
\boxed{\phi = \arctan\left(-\frac{v_0}{\omega_0 x_0}\right)}
$$

## 4. Parâmetros do Sistema

### 4.1 Período de Oscilação
O tempo necessário para completar um ciclo é:

$$
\boxed{T = \frac{2\pi}{\omega_0} = 2\pi \sqrt{\frac{m}{k}}}
$$

### 4.2 Frequência
O número de oscilações por unidade de tempo é:

$$
\boxed{f = \frac{1}{T} = \frac{\omega_0}{2\pi} = \frac{1}{2\pi} \sqrt{\frac{k}{m}}}
$$

$\omega_0 = \frac{2\pi}{T} = 2\pi f$ (a frequência independe da amplitude)

## 5. Energia do Sistema

### 5.1 Energia Cinética
$$
\boxed{E_c(t) = \frac{1}{2} m v(t)^2 = \frac{1}{2} m A^2 \omega_0^2 \sin^2(\omega_0 t + \phi)}
$$

### 5.2 Energia Potencial Elástica
$$
\boxed{E_p(t) = \frac{1}{2} k x(t)^2 = \frac{1}{2} k A^2 \cos^2(\omega_0 t + \phi)}
$$

### 5.3 Energia Mecânica Total
Seja $k = m \omega_0^2$, então:

$$
E_{total}(t) = E_c(t) + E_p(t) = \frac{1}{2} m A^2 \omega_0^2 [\sin^2(\omega_0 t + \phi) + \cos^2(\omega_0 t + \phi)]
$$

$$
\boxed{E_{T} = \frac{1}{2} m \omega_0^2 A^2 = \frac{1}{2} k A^2 = \text{const.}}
$$

Ou seja, no OHS ideal (sem amortecimento) a energia mecânica total **se conserva**.

## 6. Sistema de Equações de Primeira Ordem

Para aplicação de métodos numéricos como o **Runge-Kutta**, transformamos a EDO de segunda ordem em um sistema de duas EDOs de primeira ordem:

Definindo:
- $y_1 = x$ (posição)
- $y_2 = \dot{x} = v$ (velocidade)

O sistema fica:

$$
\begin{cases}
\dot{y}_1 = y_2 \\[10pt]
\dot{y}_2 = -\omega_0^2 y_1
\end{cases}
$$

Na forma vetorial:

$$
\boxed{\frac{d}{dt} \begin{bmatrix} y_1 \\ y_2 \end{bmatrix} = \begin{bmatrix} 0 & 1 \\ -\omega_0^2 & 0 \end{bmatrix} \begin{bmatrix} y_1 \\ y_2 \end{bmatrix}}
$$

## 7. Espaço de Fases

O **espaço de fases** é definido pelos eixos $(x, v)$, onde cada ponto representa o estado instantâneo do sistema.

### 7.1 Trajetória no Espaço de Fases
Para o oscilador harmônico simples, a trajetória é uma elipse:

$$
\boxed{\frac{x^2}{A^2} + \frac{v^2}{(\omega_0 A)^2} = 1}
$$

### 7.2 Características da Elipse
- **Semi-eixo horizontal**: $A$ (amplitude máxima)
- **Semi-eixo vertical**: $\omega_0 A$ (velocidade máxima)
- A área da elipse: $\pi A (\omega_0 A) = \pi \omega_0 A^2$ (proporcional ao quadrado da amplitude total do sistema)

## 8. Método Runge-Kutta de 4ª Ordem (RK4)

Para resolver numericamente o sistema, utilizamos o método RK4:

Dado o sistema $\dot{\mathbf{y}} = \mathbf{f}(t, \mathbf{y})$, com passo $h = \Delta t$, temos como equações principais:

$$
\begin{aligned}
\mathbf{k}_1 &= \mathbf{f}(t_n, \mathbf{y}_n) \\
\mathbf{k}_2 &= \mathbf{f}\left(t_n + \frac{h}{2}, \mathbf{y}_n + \frac{h}{2}\mathbf{k}_1\right) \\
\mathbf{k}_3 &= \mathbf{f}\left(t_n + \frac{h}{2}, \mathbf{y}_n + \frac{h}{2}\mathbf{k}_2\right) \\
\mathbf{k}_4 &= \mathbf{f}\left(t_n + h, \mathbf{y}_n + h\mathbf{k}_3\right)
\end{aligned}
$$

$$
\boxed{\mathbf{y}_{n+1} = \mathbf{y}_n + \frac{h}{6}(\mathbf{k}_1 + 2\mathbf{k}_2 + 2\mathbf{k}_3 + \mathbf{k}_4)}
$$

## 8.1. O que é o RK4?

É uma fórmula que calcula posição e velocidade no próximo instante de tempo utilizando 4 estimativas da derivada dentro do intervalo.

## 8.2. O problema

A equação $x'' + \omega_0^2 x = 0$ se torna:

$$
\begin{cases}
x' = v \\
v' = -\omega_0^2 x
\end{cases}
$$

## 8.3. A fórmula

Para avançar de $(x_n, v_n)$ para $(x_{n+1}, v_{n+1})$ com passo $h$:

$$
\boxed{(x_{n+1}, v_{n+1}) = (x_n, v_n) + \frac{h}{6}(k_1 + 2k_2 + 2k_3 + k_4)}
$$

## 8.4. As variáveis

| k | Onde calcula | Como calcula |
|---|--------------|--------------|
| $k_1$ | no início | $(v_n,\; -\omega_0^2 x_n)$ |
| $k_2$ | no meio | $\left(v_n + \frac{h}{2}k_{1v},\; -\omega_0^2\left(x_n + \frac{h}{2}k_{1x}\right)\right)$ |
| $k_3$ | no meio | $\left(v_n + \frac{h}{2}k_{2v},\; -\omega_0^2\left(x_n + \frac{h}{2}k_{2x}\right)\right)$ |
| $k_4$ | no fim | $\left(v_n + h k_{3v},\; -\omega_0^2\left(x_n + h k_{3x}\right)\right)$ |

Média ponderada (1,2,2,1) dá mais peso ao meio do intervalo

## 8.5. Exemplo numérico

Para $\omega = 1$, $h = 0.1$, partindo de $x = 1$, $v = 0$:

**Cálculo dos k's:**

$$
\begin{aligned}
k_1 &= (0,\; -1 \cdot 1) = (0, -1) \\
k_2 &= \left[0 + \frac{0.1}{2} \cdot 0,\; -1 \cdot \left(1 + \frac{0.1}{2} \cdot 0\right)\right] = (0, -1) \\
k_3 &= \left[0 + \frac{0.1}{2} \cdot 0,\; -1 \cdot \left(1 + \frac{0.1}{2} \cdot 0\right)\right] = (0, -1) \\
k_4 &= \left[0 + 0.1 \cdot 0,\; -1 \cdot \left(1 + 0.1 \cdot 0\right)\right] = (0, -1)
\end{aligned}
$$

**Próximas posição e velocidade:**

$$
\begin{aligned}
x_{n+1} &= 1 + \frac{0.1}{6} \cdot (0 + 2 \cdot 0 + 2 \cdot 0 + 0) = 1 \\
v_{n+1} &= 0 + \frac{0.1}{6} \cdot [-1 + 2 \cdot (-1) + 2 \cdot (-1) + (-1)] = -0.1
\end{aligned}
$$

## 9. Ressalvas

### 9.1 Premissas do Modelo
1. **Mola ideal**: massa desprezível, obedece à Lei de Hooke
2. **Sem amortecimento**: não há forças dissipativas (atrito, resistência do ar)
3. **Movimento unidimensional**: restrito ao eixo x
4. **Ponto material**: massa concentrada em um ponto

### 9.2 Limitações do Modelo
- Para grandes amplitudes, a Lei de Hooke pode não ser válida (regime não linear)
- Na prática, sempre há alguma força dissipativa
- A massa da mola pode ser significativa para sistemas muito leves

## 10. Generalizações

### 10.1 Oscilador Harmônico Amortecido
Adicionando uma força de amortecimento proporcional à velocidade ($F_{amort} = -b v$):

$$
\boxed{m\frac{d^2x}{dt^2} + b\frac{dx}{dt} + kx = 0}
$$

### 10.2 Oscilador Harmônico Forçado
Adicionando uma força externa periódica ($F_{ext} = F_0 \cos(\omega t)$):

$$
\boxed{m\frac{d^2x}{dt^2} + b\frac{dx}{dt} + kx = F_0 \cos(\omega t)}
$$

## Principais Equações

| Grandeza | Equação |
|----------|---------|
| Força restauradora | $F = -kx$ |
| Equação do movimento | $\frac{d^2x}{dt^2} + \omega_0^2 x = 0$ |
| Frequência angular | $\omega_0 = \sqrt{\frac{k}{m}}$ |
| Período | $T = 2\pi\sqrt{\frac{m}{k}}$ |
| Frequência linear | $f = \frac{1}{2\pi}\sqrt{\frac{k}{m}}$ |
| Solução | $x(t) = A\cos(\omega_0 t + \phi)$ |
| Amplitude | $A = \sqrt{x_0^2 + \left(\frac{v_0}{\omega_0}\right)^2}$ |
| Energia total | $E = \frac{1}{2}kA^2$ |
| Espaço de fases | $\frac{x^2}{A^2} + \frac{v^2}{(\omega_0 A)^2} = 1$ |


# **Configurações Gerais**

```python
import torch
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import pandas as pd
from scipy.fft import fft, fftfreq
```

```python
class OsciladorHarmonicoPyTorch:
    """
    Classe para resolver a equação do OHS usando PyTorch e Runge-Kutta
    """
    
    def __init__(self, massa=1.0, constante_elastica=1.0, amortecimento=0.0, device='cpu'):
        """
        Parâmetros do oscilador
        
        Args:
            massa (float): massa do sistema (kg)
            constante_elastica (float): constante elástica da mola (N/m)
            amortecimento (float): coeficiente de amortecimento (kg/s)
            device (str): dispositivo para computação ('cpu' ou 'cuda')
        """
        self.m = torch.tensor(massa, dtype=torch.float32, device=device)
        self.k = torch.tensor(constante_elastica, dtype=torch.float32, device=device)
        self.b = torch.tensor(amortecimento, dtype=torch.float32, device=device)
        self.device = device
        
        # frequência natural teórica
        self.frequencia_natural = torch.sqrt(self.k / self.m).item()
        
    def equacoes_movimento(self, estados):
        """
        Define as equações do movimento para múltiplas condições iniciais
        
        Args:
            estados: Tensor de forma (n_condicoes, 2) onde cada linha é [x, v]
        
        Returns:
            Tensor com derivadas (dx/dt, dv/dt)
        """
        x = estados[:, 0]  # posições
        v = estados[:, 1]  # velocidades
        
        dxdt = v
        dvdt = -(self.k / self.m) * x - (self.b / self.m) * v
        
        return torch.stack([dxdt, dvdt], dim=1)
    
    def runge_kutta_4(self, estados, dt):
        """
        Método Runge-Kutta de 4ª ordem para múltiplas condições iniciais
        
        Args:
            estados: Tensor de forma (n_condicoes, 2)
            dt: passo de tempo
        
        Returns:
            Tensor com novos estados após um passo
        """
        k1 = self.equacoes_movimento(estados)
        k2 = self.equacoes_movimento(estados + 0.5 * dt * k1)
        k3 = self.equacoes_movimento(estados + 0.5 * dt * k2)
        k4 = self.equacoes_movimento(estados + dt * k3)
        
        estados_novos = estados + (dt / 6.0) * (k1 + 2*k2 + 2*k3 + k4)
        
        return estados_novos
        
    def calcular_frequencias(self, posicoes, dt):
        """
        Calcula as frequências de oscilação usando FFT para cada simulação
        """
        n_condicoes = posicoes.shape[1]
        n_passos = posicoes.shape[0]
        frequencias = np.zeros(n_condicoes)
        
        for i in range(n_condicoes):
            # Aplicar janela de Hanning
            janela = np.hanning(n_passos)
            sinal_janelado = posicoes[:, i] * janela
            
            # FFT
            fft_vals = fft(sinal_janelado)
            freq_vals = fftfreq(n_passos, dt)
            
            # Pegar apenas frequências positivas
            idx_positivos = freq_vals > 0
            fft_vals_pos = np.abs(fft_vals[idx_positivos])
            freq_pos = freq_vals[idx_positivos]
            
            # Encontrar pico principal (excluindo DC)
            if len(fft_vals_pos) > 1:
                idx_principal = np.argmax(fft_vals_pos[1:]) + 1
                frequencias[i] = freq_pos[idx_principal]
            else:
                frequencias[i] = 0.0
        
        return frequencias
    
    def resolver_multi_condicoes(self, condicoes_iniciais, t_final, dt):
        """
        Resolve a EDO para múltiplas condições iniciais simultaneamente usando PyTorch
        
        Args:
            condicoes_iniciais: Tensor de forma (n_condicoes, 2) com [x0, v0]
            t_final (float): tempo final
            dt (float): passo de integração
        
        Returns:
            dict: Dicionário com tempos, posições, velocidades e energias
        """
        n_condicoes = condicoes_iniciais.shape[0]
        n_passos = int(t_final / dt) + 1
        
        # inicializa tensores para armazenar resultados
        tempos = torch.linspace(0, t_final, n_passos, device=self.device)
        posicoes = torch.zeros((n_passos, n_condicoes), device=self.device)
        velocidades = torch.zeros((n_passos, n_condicoes), device=self.device)
        
        # condições iniciais
        estados = condicoes_iniciais.clone()
        posicoes[0] = estados[:, 0]
        velocidades[0] = estados[:, 1]
        
        # integra usando Runge-Kutta
        for i in range(1, n_passos):
            estados = self.runge_kutta_4(estados, dt)
            posicoes[i] = estados[:, 0]
            velocidades[i] = estados[:, 1]
        
        # calcula energias
        energia_cinetica = 0.5 * self.m * velocidades**2
        energia_potencial = 0.5 * self.k * posicoes**2
        energia_mecanica = energia_cinetica + energia_potencial
        
        # calcula frequências via FFT para cada condição
        frequencias = self.calcular_frequencias(posicoes.cpu().numpy(), dt)
        
        return {
            'tempo': tempos.cpu().numpy(),
            'posicao': posicoes.cpu().numpy(),
            'velocidade': velocidades.cpu().numpy(),
            'energia_cinetica': energia_cinetica.cpu().numpy(),
            'energia_potencial': energia_potencial.cpu().numpy(),
            'energia_mecanica': energia_mecanica.cpu().numpy(),
            'frequencias': frequencias,
            'condicoes_iniciais': condicoes_iniciais.cpu().numpy()
        }
    
    def gerar_base_dados(self, condicoes_iniciais, t_final, dt):
        """
        Gera uma base de dados completa com as soluções
        
        Args:
            condicoes_iniciais: Tensor com condições iniciais
            t_final (float): tempo final
            dt (float): passo de integração
        
        Returns:
            pd.DataFrame: DataFrame com todos os dados
        """
        # resolve para todas as condições iniciais
        solucao = self.resolver_multi_condicoes(condicoes_iniciais, t_final, dt)
        
        # cria lista para armazenar dados
        dados = []
        
        n_condicoes = condicoes_iniciais.shape[0]
        n_passos = len(solucao['tempo'])
        
        for i in range(n_condicoes):
            for j in range(n_passos):
                dados.append({
                    'simulacao_id': i,
                    'x0': solucao['condicoes_iniciais'][i, 0],
                    'v0': solucao['condicoes_iniciais'][i, 1],
                    'tempo': solucao['tempo'][j],
                    'posicao': solucao['posicao'][j, i],
                    'velocidade': solucao['velocidade'][j, i],
                    'energia_cinetica': solucao['energia_cinetica'][j, i],
                    'energia_potencial': solucao['energia_potencial'][j, i],
                    'energia_mecanica': solucao['energia_mecanica'][j, i],
                    'frequencia_sistema': solucao['frequencias'][i]
                })
        
        df = pd.DataFrame(dados)
        df.to_csv('dados_ohs_pytorch.csv', index=False)
        print(f"\nBase de dados salva em 'dados_ohs_pytorch.csv'")
        print(f"Total de Registros: {len(df)}")
        
        return df, solucao
```

```python

def criar_visualizacoes_plotly(solucao, idx_simulacao=0):
    """
    Cria visualizações interativas usando Plotly
    """
    tempo = solucao['tempo']
    posicao = solucao['posicao'][:, idx_simulacao]
    velocidade = solucao['velocidade'][:, idx_simulacao]
    energia_cinetica = solucao['energia_cinetica'][:, idx_simulacao]
    energia_potencial = solucao['energia_potencial'][:, idx_simulacao]
    energia_mecanica = solucao['energia_mecanica'][:, idx_simulacao]
    
    x0 = solucao['condicoes_iniciais'][idx_simulacao, 0]
    v0 = solucao['condicoes_iniciais'][idx_simulacao, 1]
    freq = solucao['frequencias'][idx_simulacao]
    
    # Criar subplots
    fig = make_subplots(
        rows=2, cols=2,
        subplot_titles=('Espaço de Fases', 'Posição vs Tempo',
                       'Energias vs Tempo', 'Velocidade vs Tempo'),
        specs=[[{'type': 'scatter'}, {'type': 'scatter'}],
               [{'type': 'scatter'}, {'type': 'scatter'}]]
    )
    
    # Espaço de fases
    fig.add_trace(
        go.Scatter(x=posicao, y=velocidade,
                  mode='lines',
                  line=dict(color='blue', width=2),
                  name='Espaço de Fases',
                  hovertemplate='Posição: %{x:.3f} m<br>Velocidade: %{y:.3f} m/s<extra></extra>'),
        row=1, col=1
    )
    
    # Posição vs Tempo
    fig.add_trace(
        go.Scatter(x=tempo, y=posicao,
                  mode='lines',
                  line=dict(color='red', width=2),
                  name='Posição',
                  hovertemplate='Tempo: %{x:.3f} s<br>Posição: %{y:.3f} m<extra></extra>'),
        row=1, col=2
    )
    
    # Energias vs Tempo
    fig.add_trace(
        go.Scatter(x=tempo, y=energia_cinetica,
                  mode='lines',
                  line=dict(color='green', width=2),
                  name='Energia Cinética'),
        row=2, col=1
    )
    
    fig.add_trace(
        go.Scatter(x=tempo, y=energia_potencial,
                  mode='lines',
                  line=dict(color='orange', width=2),
                  name='Energia Potencial'),
        row=2, col=1
    )
    
    fig.add_trace(
        go.Scatter(x=tempo, y=energia_mecanica,
                  mode='lines',
                  line=dict(color='red', width=3, dash='dash'),
                  name='Energia Mecânica'),
        row=2, col=1
    )
    
    # Velocidade vs Tempo
    fig.add_trace(
        go.Scatter(x=tempo, y=velocidade,
                  mode='lines',
                  line=dict(color='purple', width=2),
                  name='Velocidade',
                  hovertemplate='Tempo: %{x:.3f} s<br>Velocidade: %{y:.3f} m/s<extra></extra>'),
        row=2, col=2
    )
    
    # Atualizar layout
    fig.update_xaxes(title_text="Posição (m)", row=1, col=1)
    fig.update_yaxes(title_text="Velocidade (m/s)", row=1, col=1)
    fig.update_xaxes(title_text="Tempo (s)", row=1, col=2)
    fig.update_yaxes(title_text="Posição (m)", row=1, col=2)
    fig.update_xaxes(title_text="Tempo (s)", row=2, col=1)
    fig.update_yaxes(title_text="Energia (J)", row=2, col=1)
    fig.update_xaxes(title_text="Tempo (s)", row=2, col=2)
    fig.update_yaxes(title_text="Velocidade (m/s)", row=2, col=2)
    
    # Título principal
    fig.update_layout(
        title_text=f"Oscilador Harmônico Simples - x0={x0:.2f} m, v0={v0:.2f} m/s, f={freq:.3f} Hz",
        height=800,
        showlegend=True,
        hovermode='closest'
    )
    
    return fig
```

```python
def criar_animacao_espaco_fases(solucao, idx_simulacao=0):
    """
    Cria uma animação interativa do espaço de fases
    """
    tempo = solucao['tempo']
    posicao = solucao['posicao'][:, idx_simulacao]
    velocidade = solucao['velocidade'][:, idx_simulacao]
    x0 = solucao['condicoes_iniciais'][idx_simulacao, 0]
    v0 = solucao['condicoes_iniciais'][idx_simulacao, 1]
    
    fig = go.Figure()
    
    # Adicionar a trajetória completa
    fig.add_trace(
        go.Scatter(x=posicao, y=velocidade,
                  mode='lines',
                  line=dict(color='lightgray', width=2),
                  name='Trajetória',
                  showlegend=True)
    )
    
    # Adicionar ponto animado
    fig.add_trace(
        go.Scatter(x=[posicao[0]], y=[velocidade[0]],
                  mode='markers',
                  marker=dict(size=10, color='red'),
                  name='Posição atual',
                  showlegend=True)
    )
    
    # Criar frames para animação
    frames = []
    for i in range(0, len(tempo), max(1, len(tempo)//100)):  # Usar 100 frames
        frames.append(
            go.Frame(
                data=[
                    go.Scatter(x=posicao[:i+1], y=velocidade[:i+1],
                              mode='lines', line=dict(color='lightgray')),
                    go.Scatter(x=[posicao[i]], y=[velocidade[i]],
                              mode='markers', marker=dict(size=10, color='red'))
                ],
                name=f'frame{i}'
            )
        )
    
    fig.frames = frames
    
    # Configurar layout
    fig.update_layout(
        title=f"Animação do Espaço de Fases - x0={x0:.2f} m, v0={v0:.2f} m/s",
        xaxis_title="Posição (m)",
        yaxis_title="Velocidade (m/s)",
        updatemenus=[dict(
            type="buttons",
            buttons=[dict(label="Play",
                         method="animate",
                         args=[None, {"frame": {"duration": 50, "redraw": True},
                                     "fromcurrent": True}])],
            showactive=False,
            x=0.1,
            y=0
        )],
        width=800,
        height=600
    )
    
    return fig
```

```python
def criar_heatmap_energias(solucao):
    """
    Cria um heatmap interativo das energias no espaço de fases
    """
    # Selecionar apenas alguns pontos para o heatmap
    n_condicoes = solucao['posicao'].shape[1]
    
    # Calcular energia máxima para cada simulação
    energias_max = np.max(solucao['energia_mecanica'], axis=0)
    
    # Criar grid para o heatmap
    x0_vals = solucao['condicoes_iniciais'][:, 0]
    v0_vals = solucao['condicoes_iniciais'][:, 1]
    
    fig = go.Figure(data=go.Scatter(
        x=x0_vals,
        y=v0_vals,
        mode='markers',
        marker=dict(
            size=10,
            color=energias_max,
            colorscale='Viridis',
            showscale=True,
            colorbar=dict(title="Energia Máxima (J)")
        ),
        text=[f"x0={x0:.2f}, v0={v0:.2f}<br>E_max={e:.2f} J" 
              for x0, v0, e in zip(x0_vals, v0_vals, energias_max)],
        hovertemplate='%{text}<extra></extra>'
    ))
    
    fig.update_layout(
        title="Energia Máxima no Espaço de Condições Iniciais",
        xaxis_title="Posição Inicial (m)",
        yaxis_title="Velocidade Inicial (m/s)",
        width=800,
        height=600
    )
    
    return fig
```

```python
def main():
    """
    Função principal
    """
    # Verificar se CUDA está disponível
    device = 'cuda' if torch.cuda.is_available() else 'cpu'
    print(f"Usando dispositivo: {device}")
    
    # Criar oscilador
    oscilador = OsciladorHarmonicoPyTorch(
        massa=1.0, 
        constante_elastica=1.0, 
        device=device
    )
    
    # Definir parâmetros de simulação
    t_final = 20.0  # 20 segundos
    dt = 0.01  # passo de 0.01 segundos
    
    # Criar condições iniciais usando PyTorch
    print("\n=== CRIANDO CONDIÇÕES INICIAIS ===")
    
    # Gerar grid de condições iniciais
    x0_range = torch.linspace(-2, 2, 15, device=device)
    v0_range = torch.linspace(-2, 2, 15, device=device)
    
    # Criar meshgrid
    x0_grid, v0_grid = torch.meshgrid(x0_range, v0_range, indexing='ij')
    condicoes_iniciais = torch.stack([x0_grid.flatten(), v0_grid.flatten()], dim=1)
    
    print(f"Número de simulações: {condicoes_iniciais.shape[0]}")
    print(f"Formato dos tensores: {condicoes_iniciais.shape}")
    
    # Gerar base de dados
    print("\n=== GERANDO BASE DE DADOS ===")
    df, solucao = oscilador.gerar_base_dados(condicoes_iniciais, t_final, dt)
    
    # Criar visualizações
    print("\n=== CRIANDO VISUALIZAÇÕES ===")
    
    # Visualização para uma simulação específica
    idx_sim = 112  # Selecionar uma simulação central
    fig1 = criar_visualizacoes_plotly(solucao, idx_sim)
    fig1.show()
    
    # Animação do espaço de fases
    fig2 = criar_animacao_espaco_fases(solucao, idx_sim)
    fig2.show()
    
    # Heatmap de energias
    fig3 = criar_heatmap_energias(solucao)
    fig3.show()
    
    # Criar visualização 3D do espaço de fases
    fig4 = go.Figure()
    
    # Plotar várias trajetórias em 3D
    n_trajetorias = min(20, solucao['posicao'].shape[1])
    for i in range(n_trajetorias):
        fig4.add_trace(go.Scatter3d(
            x=solucao['posicao'][:, i],
            y=solucao['velocidade'][:, i],
            z=solucao['tempo'],
            mode='lines',
            line=dict(width=2),
            name=f'Sim {i}',
            opacity=0.7
        ))
    
    fig4.update_layout(
        title="Espaço de Fases 3D (Posição vs Velocidade vs Tempo)",
        scene=dict(
            xaxis_title="Posição (m)",
            yaxis_title="Velocidade (m/s)",
            zaxis_title="Tempo (s)"
        ),
        width=1000,
        height=800
    )
    fig4.show()
    
    # Estatísticas finais
    print("\n=== ESTATÍSTICAS DA BASE DE DADOS ===")
    print(f"Total de simulações: {df['simulacao_id'].nunique()}")
    print(f"Total de pontos: {len(df)}")
    print(f"Frequência natural teórica: {oscilador.frequencia_natural/(2*np.pi):.3f} Hz")
    print(f"Média das frequências calculadas: {solucao['frequencias'].mean():.3f} Hz")
    print(f"Desvio padrão das frequências: {solucao['frequencias'].std():.3f} Hz")
    
    print("\n=== AMOSTRA DA BASE DE DADOS ===")
    print(df.head(20))
    
    return df, solucao

if __name__ == "__main__":
    df, solucao = main()
```

```python
import torch
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import pandas as pd
from scipy.fft import fft, fftfreq
from scipy.signal import find_peaks
import locale

# Configurar locale para formato brasileiro
try:
    locale.setlocale(locale.LC_NUMERIC, 'pt_BR.UTF-8')
except:
    try:
        locale.setlocale(locale.LC_NUMERIC, 'Portuguese_Brazil.1252')
    except:
        print("Locale brasileiro não disponível, usando formato padrão")
        locale.setlocale(locale.LC_NUMERIC, 'C')

class OsciladorHarmonicoPyTorch:
    """
    Classe para resolver a equação do oscilador harmônico simples usando PyTorch e Runge-Kutta
    """
    
    def __init__(self, massa=1.0, constante_elastica=1.0, amortecimento=0.0, device='cpu'):
        """
        Parâmetros do oscilador
        
        Args:
            massa (float): massa do sistema (kg)
            constante_elastica (float): constante elástica da mola (N/m)
            amortecimento (float): coeficiente de amortecimento (kg/s)
            device (str): dispositivo para computação ('cpu' ou 'cuda')
        """
        self.m = torch.tensor(massa, dtype=torch.float32, device=device)
        self.k = torch.tensor(constante_elastica, dtype=torch.float32, device=device)
        self.b = torch.tensor(amortecimento, dtype=torch.float32, device=device)
        self.device = device
        
        # Frequência natural teórica (em Hz)
        self.frequencia_natural_hz = torch.sqrt(self.k / self.m).item() / (2 * np.pi)
        self.frequencia_angular = torch.sqrt(self.k / self.m).item()
        
    def equacoes_movimento(self, estados):
        """
        Define as equações do movimento para múltiplas condições iniciais
        """
        x = estados[:, 0]
        v = estados[:, 1]
        
        dxdt = v
        dvdt = -(self.k / self.m) * x - (self.b / self.m) * v
        
        return torch.stack([dxdt, dvdt], dim=1)
    
    def runge_kutta_4(self, estados, dt):
        """
        Método Runge-Kutta de 4ª ordem para múltiplas condições iniciais
        """
        k1 = self.equacoes_movimento(estados)
        k2 = self.equacoes_movimento(estados + 0.5 * dt * k1)
        k3 = self.equacoes_movimento(estados + 0.5 * dt * k2)
        k4 = self.equacoes_movimento(estados + dt * k3)
        
        estados_novos = estados + (dt / 6.0) * (k1 + 2*k2 + 2*k3 + k4)
        
        return estados_novos
    
    def resolver_multi_condicoes(self, condicoes_iniciais, t_final, dt):
        """
        Resolve a EDO para múltiplas condições iniciais simultaneamente
        """
        n_condicoes = condicoes_iniciais.shape[0]
        n_passos = int(t_final / dt) + 1
        
        tempos = torch.linspace(0, t_final, n_passos, device=self.device)
        posicoes = torch.zeros((n_passos, n_condicoes), device=self.device)
        velocidades = torch.zeros((n_passos, n_condicoes), device=self.device)
        
        estados = condicoes_iniciais.clone()
        posicoes[0] = estados[:, 0]
        velocidades[0] = estados[:, 1]
        
        for i in range(1, n_passos):
            estados = self.runge_kutta_4(estados, dt)
            posicoes[i] = estados[:, 0]
            velocidades[i] = estados[:, 1]
        
        # Calcular energias
        energia_cinetica = 0.5 * self.m * velocidades**2
        energia_potencial = 0.5 * self.k * posicoes**2
        energia_mecanica = energia_cinetica + energia_potencial
        
        # Calcular amplitudes máximas
        amplitudes_max = torch.max(torch.abs(posicoes), dim=0)[0].cpu().numpy()
        
        # Calcular frequências com maior precisão
        frequencias, periodos = self.calcular_frequencias_preciso(posicoes.cpu().numpy(), dt)
        
        return {
            'tempo': tempos.cpu().numpy(),
            'posicao': posicoes.cpu().numpy(),
            'velocidade': velocidades.cpu().numpy(),
            'energia_cinetica': energia_cinetica.cpu().numpy(),
            'energia_potencial': energia_potencial.cpu().numpy(),
            'energia_mecanica': energia_mecanica.cpu().numpy(),
            'frequencias': frequencias,
            'periodos': periodos,
            'amplitudes': amplitudes_max,
            'condicoes_iniciais': condicoes_iniciais.cpu().numpy()
        }
    
    def calcular_frequencias_preciso(self, posicoes, dt):
        """
        Calcula frequências com maior precisão usando múltiplos métodos
        """
        n_condicoes = posicoes.shape[1]
        n_passos = posicoes.shape[0]
        frequencias = np.zeros(n_condicoes)
        periodos = np.zeros(n_condicoes)
        
        for i in range(n_condicoes):
            sinal = posicoes[:, i]
            
            # Método 1: FFT com janela e zero-padding para melhor resolução
            n_fft = n_passos * 4  # Zero-padding para melhor resolução
            janela = np.hanning(n_passos)
            sinal_janelado = sinal * janela
            
            fft_vals = fft(sinal_janelado, n=n_fft)
            freq_vals = fftfreq(n_fft, dt)
            
            idx_positivos = freq_vals > 0
            fft_vals_pos = np.abs(fft_vals[idx_positivos])
            freq_pos = freq_vals[idx_positivos]
            
            # Encontrar pico principal (excluindo DC)
            if len(fft_vals_pos) > 1:
                # Usar apenas frequências acima de 0.01 Hz para evitar ruído
                mask = freq_pos > 0.01
                if np.any(mask):
                    freq_filtradas = freq_pos[mask]
                    fft_filtradas = fft_vals_pos[mask]
                    idx_principal = np.argmax(fft_filtradas)
                    freq_fft = freq_filtradas[idx_principal]
                    
                    # Método 2: Contagem de cruzamentos por zero para verificar
                    cruzamentos_zero = np.where(np.diff(np.signbit(sinal)))[0]
                    if len(cruzamentos_zero) > 10:
                        # Calcular período médio entre cruzamentos
                        periodos_zero = np.diff(cruzamentos_zero) * dt
                        periodo_medio = np.mean(periodos_zero[periodos_zero > 0])
                        freq_cruzamento = 1.0 / (2.0 * periodo_medio)  # Fator 2 pois cada período tem 2 cruzamentos
                        
                        # Usar média ponderada entre os dois métodos
                        frequencias[i] = 0.7 * freq_fft + 0.3 * freq_cruzamento
                        periodos[i] = 1.0 / frequencias[i]
                    else:
                        frequencias[i] = freq_fft
                        periodos[i] = 1.0 / freq_fft
                else:
                    frequencias[i] = self.frequencia_natural_hz
                    periodos[i] = 1.0 / self.frequencia_natural_hz
            else:
                frequencias[i] = self.frequencia_natural_hz
                periodos[i] = 1.0 / self.frequencia_natural_hz
        
        return frequencias, periodos
    
    def formatar_numero_brasil(self, numero):
        """
        Formata número no padrão brasileiro
        """
        if isinstance(numero, (int, float)):
            if abs(numero) < 1000:
                return f"{numero:.3f}".replace('.', ',')
            else:
                partes = f"{numero:.3f}".split('.')
                parte_inteira = '{:,.0f}'.format(int(partes[0])).replace(',', '.')
                return f"{parte_inteira},{partes[1]}"
        return str(numero)
    
    def gerar_base_dados(self, condicoes_iniciais, t_final, dt):
        """
        Gera uma base de dados completa com as soluções no formato brasileiro
        """
        solucao = self.resolver_multi_condicoes(condicoes_iniciais, t_final, dt)
        
        dados = []
        n_condicoes = condicoes_iniciais.shape[0]
        n_passos = len(solucao['tempo'])
        
        for i in range(n_condicoes):
            for j in range(n_passos):
                dados.append({
                    'simulacao_id': i,
                    'x0': self.formatar_numero_brasil(solucao['condicoes_iniciais'][i, 0]),
                    'v0': self.formatar_numero_brasil(solucao['condicoes_iniciais'][i, 1]),
                    'amplitude_max': self.formatar_numero_brasil(solucao['amplitudes'][i]),
                    'frequencia_hz': self.formatar_numero_brasil(solucao['frequencias'][i]),
                    'periodo_s': self.formatar_numero_brasil(solucao['periodos'][i]),
                    'tempo': self.formatar_numero_brasil(solucao['tempo'][j]),
                    'posicao': self.formatar_numero_brasil(solucao['posicao'][j, i]),
                    'velocidade': self.formatar_numero_brasil(solucao['velocidade'][j, i]),
                    'energia_cinetica': self.formatar_numero_brasil(solucao['energia_cinetica'][j, i]),
                    'energia_potencial': self.formatar_numero_brasil(solucao['energia_potencial'][j, i]),
                    'energia_mecanica': self.formatar_numero_brasil(solucao['energia_mecanica'][j, i])
                })
        
        df = pd.DataFrame(dados)
        df.to_csv('dados_oscilador_pytorch.csv', index=False, sep=';', encoding='utf-8')
        print(f"\nBase de dados salva em 'dados_oscilador_pytorch.csv'")
        print(f"Total de registros: {len(df):,}".replace(',', '.'))
        
        return df, solucao

def selecionar_trajetorias_representativas(solucao, n_trajetorias=20):
    """
    Seleciona trajetórias representativas baseado em diferentes características físicas
    """
    n_total = solucao['frequencias'].shape[0]
    amplitudes = solucao['amplitudes']
    frequencias = solucao['frequencias']
    cond_iniciais = solucao['condicoes_iniciais']
    
    # Critérios para seleção diversificada:
    # 1. Diferentes amplitudes
    # 2. Diferentes energias
    # 3. Diferentes condições iniciais (fases)
    
    # Normalizar amplitudes para seleção
    amplitudes_norm = amplitudes / amplitudes.max() if amplitudes.max() > 0 else amplitudes
    
    # Criar categorias baseadas na amplitude
    categorias = []
    for i in range(n_total):
        # Classificar por amplitude (pequena, média, grande)
        if amplitudes_norm[i] < 0.33:
            cat_amp = 0
        elif amplitudes_norm[i] < 0.66:
            cat_amp = 1
        else:
            cat_amp = 2
        
        # Classificar por fase (baseado na razão x0/v0)
        x0, v0 = cond_iniciais[i]
        if abs(x0) > abs(v0):
            cat_fase = 0  # Dominância de posição
        elif abs(v0) > abs(x0):
            cat_fase = 1  # Dominância de velocidade
        else:
            cat_fase = 2  # Equilibrado
        
        categorias.append((cat_amp, cat_fase))
    
    # Selecionar trajetórias representativas de cada categoria
    indices_selecionados = []
    categorias_unicas = list(set(categorias))
    
    # Garantir que pegamos de todas as categorias
    for cat in categorias_unicas:
        indices_cat = [i for i, c in enumerate(categorias) if c == cat]
        # Selecionar até ceil(n_trajetorias / len(categorias_unicas)) de cada categoria
        n_por_cat = max(1, int(np.ceil(n_trajetorias / len(categorias_unicas))))
        indices_selecionados.extend(indices_cat[:n_por_cat])
    
    # Se ainda não temos o suficiente, completar com as de maior energia
    if len(indices_selecionados) < n_trajetorias:
        energias_totais = solucao['energia_mecanica'][-1]  # Energia final
        indices_restantes = [i for i in range(n_total) if i not in indices_selecionados]
        indices_restantes.sort(key=lambda i: energias_totais[i], reverse=True)
        indices_selecionados.extend(indices_restantes[:n_trajetorias - len(indices_selecionados)])
    
    # Ordenar por frequência para melhor visualização
    indices_selecionados.sort(key=lambda i: frequencias[i])
    
    return indices_selecionados[:n_trajetorias]

def criar_visualizacao_3d_com_caracteristicas(solucao, indices_selecionados):
    """
    Cria visualização 3D com cada trajetória rotulada por suas características específicas
    """
    fig = go.Figure()
    
    # Gerar paleta de cores baseada nas frequências
    freqs_selecionadas = solucao['frequencias'][indices_selecionados]
    freq_min = freqs_selecionadas.min()
    freq_max = freqs_selecionadas.max()
    
    for idx in indices_selecionados:
        # Obter características desta simulação
        freq = solucao['frequencias'][idx]
        amplitude = solucao['amplitudes'][idx]
        x0, v0 = solucao['condicoes_iniciais'][idx]
        energia_total = solucao['energia_mecanica'][-1, idx]
        
        # Calcular energia total teórica: E = (1/2)*k*A²
        energia_teorica = 0.5 * 1.0 * amplitude**2  # k=1.0
        
        # Criar rótulo descritivo com as características
        rotulo = (f"f = {freq:.4f} Hz<br>" +
                 f"A = {amplitude:.3f} m<br>" +
                 f"E = {energia_total:.3f} J<br>" +
                 f"x₀ = {x0:.3f} m, v₀ = {v0:.3f} m/s")
        
        # Cor baseada na frequência (escala de cores)
        cor_normalizada = (freq - freq_min) / (freq_max - freq_min) if freq_max > freq_min else 0.5
        
        # Adicionar trajetória
        fig.add_trace(go.Scatter3d(
            x=solucao['posicao'][:, idx],
            y=solucao['velocidade'][:, idx],
            z=solucao['tempo'],
            mode='lines',
            line=dict(
                width=2,
                color=f'rgb({int(50 + 205*cor_normalizada)}, {int(50 + 150*(1-cor_normalizada))}, {int(200 - 100*cor_normalizada)})'
            ),
            name=f'{freq:.4f} Hz',
            legendgroup=f'freq_{freq:.4f}',
            showlegend=True,
            hovertemplate=rotulo + '<br>Posição: %{x:.3f} m<br>Velocidade: %{y:.3f} m/s<br>Tempo: %{z:.1f} s<extra></extra>'
        ))
    
    # Calcular estatísticas das frequências
    freq_media = freqs_selecionadas.mean()
    freq_std = freqs_selecionadas.std()
    freq_teorica = solucao['frequencias'].mean()  # Média de todas as simulações
    
    fig.update_layout(
        title=(
            f"Espaço de Fases 3D - Trajetórias Representativas<br>" +
            f"<sup>Frequência teórica: {freq_teorica:.4f} Hz | "
            f"Média das selecionadas: {freq_media:.4f} ± {freq_std:.4f} Hz</sup>"
        ),
        scene=dict(
            xaxis_title="Posição (m)",
            yaxis_title="Velocidade (m/s)",
            zaxis_title="Tempo (s)",
            camera=dict(
                eye=dict(x=1.5, y=1.5, z=1.2),
                up=dict(x=0, y=0, z=1)
            )
        ),
        width=1200,
        height=800,
        legend=dict(
            title="Frequência de Oscilação (Hz)",
            yanchor="top",
            y=0.99,
            xanchor="left",
            x=1.02,
            bgcolor="rgba(255, 255, 255, 0.9)",
            bordercolor="Black",
            borderwidth=1,
            font=dict(size=10)
        ),
        hoverlabel=dict(
            bgcolor="white",
            font_size=12,
            font_family="Arial"
        )
    )
    
    return fig

def criar_visualizacao_2d_comparativa(solucao, indices_selecionados):
    """
    Cria visualização 2D comparativa das trajetórias selecionadas
    """
    n_trajetorias = len(indices_selecionados)
    n_cols = min(4, n_trajetorias)
    n_rows = (n_trajetorias + n_cols - 1) // n_cols
    
    fig = make_subplots(
        rows=n_rows, cols=n_cols,
        subplot_titles=[f"f = {solucao['frequencias'][idx]:.4f} Hz<br>A = {solucao['amplitudes'][idx]:.3f} m" 
                        for idx in indices_selecionados],
        horizontal_spacing=0.1,
        vertical_spacing=0.12
    )
    
    for i, idx in enumerate(indices_selecionados):
        row = i // n_cols + 1
        col = i % n_cols + 1
        
        # Espaço de fases
        fig.add_trace(
            go.Scatter(
                x=solucao['posicao'][:, idx],
                y=solucao['velocidade'][:, idx],
                mode='lines',
                line=dict(color='blue', width=1.5),
                name=f'Sim {idx}',
                showlegend=False,
                hovertemplate='Posição: %{x:.3f} m<br>Velocidade: %{y:.3f} m/s<extra></extra>'
            ),
            row=row, col=col
        )
        
        # Adicionar ponto inicial
        fig.add_trace(
            go.Scatter(
                x=[solucao['posicao'][0, idx]],
                y=[solucao['velocidade'][0, idx]],
                mode='markers',
                marker=dict(color='red', size=8, symbol='circle'),
                showlegend=False,
                hovertemplate='<b>Condição Inicial</b><br>x₀ = %{x:.3f} m<br>v₀ = %{y:.3f} m/s<extra></extra>'
            ),
            row=row, col=col
        )
        
        fig.update_xaxes(title_text="Posição (m)", row=row, col=col)
        fig.update_yaxes(title_text="Velocidade (m/s)", row=row, col=col)
    
    fig.update_layout(
        title_text="Espaço de Fases - Comparação entre Diferentes Trajetórias",
        height=300 * n_rows,
        width=400 * n_cols,
        showlegend=False
    )
    
    return fig

def main():
    """
    Função principal
    """
    device = 'cuda' if torch.cuda.is_available() else 'cpu'
    print(f"Usando dispositivo: {device}")
    print(f"Frequência natural teórica: {1/(2*np.pi):.4f} Hz (para k=1, m=1)")
    
    # Criar oscilador
    oscilador = OsciladorHarmonicoPyTorch(
        massa=1.0, 
        constante_elastica=1.0, 
        device=device
    )
    
    # Parâmetros de simulação
    t_final = 30.0  # Aumentado para melhor resolução de frequência
    dt = 0.01
    
    # Criar grid completo de condições iniciais
    print("\n=== CRIANDO CONDIÇÕES INICIAIS ===")
    
    # Grid mais fino para melhor cobertura do espaço de fases
    x0_range = torch.linspace(-2.5, 2.5, 20, device=device)
    v0_range = torch.linspace(-2.5, 2.5, 20, device=device)
    
    x0_grid, v0_grid = torch.meshgrid(x0_range, v0_range, indexing='ij')
    condicoes_iniciais = torch.stack([x0_grid.flatten(), v0_grid.flatten()], dim=1)
    
    n_simulacoes = condicoes_iniciais.shape[0]
    print(f"Número total de simulações: {n_simulacoes}")
    print(f"Grid: {len(x0_range)} x {len(v0_range)} = {n_simulacoes}")
    
    # Gerar base de dados
    print("\n=== GERANDO BASE DE DADOS ===")
    df, solucao = oscilador.gerar_base_dados(condicoes_iniciais, t_final, dt)
    
    # Análise das frequências
    print("\n=== ANÁLISE DE FREQUÊNCIAS ===")
    frequencias = solucao['frequencias']
    print(f"Frequência natural teórica: {oscilador.frequencia_natural_hz:.4f} Hz")
    print(f"Média das frequências calculadas: {frequencias.mean():.4f} Hz")
    print(f"Desvio padrão: {frequencias.std():.6f} Hz")
    print(f"Mínimo: {frequencias.min():.6f} Hz")
    print(f"Máximo: {frequencias.max():.6f} Hz")
    
    # Selecionar trajetórias representativas
    print("\n=== SELECIONANDO TRAJETÓRIAS REPRESENTATIVAS ===")
    indices_selecionados = selecionar_trajetorias_representativas(solucao, n_trajetorias=20)
    
    print(f"Selecionadas {len(indices_selecionados)} trajetórias:")
    for i, idx in enumerate(indices_selecionados[:10]):  # Mostrar primeiras 10
        x0, v0 = solucao['condicoes_iniciais'][idx]
        freq = solucao['frequencias'][idx]
        amp = solucao['amplitudes'][idx]
        print(f"  {i+1:2d}. Simulação {idx:3d}: f={freq:.4f} Hz, A={amp:.3f} m, x0={x0:.3f}, v0={v0:.3f}")
    
    # Criar visualizações
    print("\n=== CRIANDO VISUALIZAÇÕES ===")
    
    # Visualização 3D com características individuais
    fig3d = criar_visualizacao_3d_com_caracteristicas(solucao, indices_selecionados)
    fig3d.show()
    
    # Visualização 2D comparativa
    fig2d = criar_visualizacao_2d_comparativa(solucao, indices_selecionados)
    fig2d.show()
    
    # Heatmap de energias com informações de frequência
    fig_heat = criar_heatmap_energias(solucao)
    fig_heat.show()
    
    # Estatísticas finais
    print("\n=== ESTATÍSTICAS DA BASE DE DADOS ===")
    print(f"Total de simulações: {df['simulacao_id'].nunique()}")
    print(f"Total de pontos: {len(df)}")
    print(f"Amplitude máxima: {solucao['amplitudes'].max():.3f} m")
    print(f"Energia máxima: {solucao['energia_mecanica'].max():.3f} J")
    
    return df, solucao, indices_selecionados

def criar_heatmap_energias(solucao):
    """
    Cria um heatmap interativo das energias no espaço de fases
    """
    energias_max = np.max(solucao['energia_mecanica'], axis=0)
    x0_vals = solucao['condicoes_iniciais'][:, 0]
    v0_vals = solucao['condicoes_iniciais'][:, 1]
    freqs = solucao['frequencias']
    
    fig = go.Figure(data=go.Scatter(
        x=x0_vals,
        y=v0_vals,
        mode='markers',
        marker=dict(
            size=12,
            color=energias_max,
            colorscale='Viridis',
            showscale=True,
            colorbar=dict(title="Energia Máxima (J)"),
            line=dict(width=1, color='black')
        ),
        text=[f"<b>Simulação {i}</b><br>" +
              f"x₀ = {x0:.3f} m<br>" +
              f"v₀ = {v0:.3f} m/s<br>" +
              f"f = {freq:.4f} Hz<br>" +
              f"E_max = {e:.3f} J"
              for i, (x0, v0, e, freq) in enumerate(zip(x0_vals, v0_vals, energias_max, freqs))],
        hovertemplate='%{text}<extra></extra>'
    ))
    
    fig.update_layout(
        title="Mapa de Energia no Espaço de Condições Iniciais",
        xaxis_title="Posição Inicial (m)",
        yaxis_title="Velocidade Inicial (m/s)",
        width=900,
        height=700,
        hoverlabel=dict(
            bgcolor="white",
            font_size=12
        )
    )
    
    return fig

if __name__ == "__main__":
    df, solucao, indices_selecionados = main()
```
